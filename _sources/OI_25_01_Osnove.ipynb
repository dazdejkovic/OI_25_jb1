# Deo I: Osnove programiranja u Pajtonu

U ovom delu udžbenika uvode se osnovni koncepti programiranja u programskom
jeziku Pajton. Poseban naglasak stavljen je na razumevanje strukture programa,
osnovnih tipova podataka, kontrolnih struktura i funkcija, kao i na osnovne
elemente objektno orijentisanog programiranja.

**Ishodi učenja.** Nakon proučavanja ovog dela student će moći da:

- napiše jednostavne programe u Pajtonu koristeći osnovne kontrolne strukture,
- koristi osnovne tipove i strukture podataka u rešavanju problema,
- definiše i koristi funkcije u programima,
- razume i implementira jednostavne klase u okviru objektno orijentisanog pristupa.
