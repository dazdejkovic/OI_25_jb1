# Osnove Pajtona



<br>

**Ciljevi učenja** u ovom delu udžbenika:

- Razumevanje osnovnih koncepata programiranja u Pajtonu (promenljive, tipovi podataka, operatori).
- Upoznavanje sa strukturom i sintaksom Pajton jezika.
- Razvijanje sposobnosti samostalnog pisanja jednostavnih Pajton programa.

**Ishodi učenja**. Nakon ovog dela kursa student će biti sposoban da:
- piše osnovne Pajton programe koristeći promenljive, petlje, uslove i funkcije,
- razlikuje glavne tipove podataka i pravilno ih koristi,
- razume i primenjuje osnovne principe programskog razmišljanja i algoritamske logike.
