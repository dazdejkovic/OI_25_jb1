## Moduli


Zamisli da želiš da izračunaš vrednosti kao što su $𝑒^{2.5}$ ili $\ln 3.45$.

Pajton bez ikakvih dodataka ume da sabira, množi, ili radi sa celim i realnim brojevima — ali ne zna sam od sebe šta su eksponencijalna funkcija, prirodni logaritam, trigonometrijske funkcije i slično. Drugim rečima: osnovni Python je namerno “mršav”.

Da bi znao da izračuna
```python
e**2.5
ln(3.45)
```
Pajtonu je potrebna dodatna matematika. Tu na scenu stupaju moduli.

Pajton je programski jezik otvorenog koda. Kao takav, postoji puno funkcionalnosti koji su korisnici širom sveta razvili i učinili dostupnim svima. 
Module shvatite kao biblioteke u kojima se nalazi niz funkcija koje će nam olakšati rad. Glavna prednost programa kao što su Pajton ili R je veliki broj ovakvih modula (biblioteka) i za nas je važno da naučimo da ih uvezemo i koristimo po potrebi.  


### Komanda `import`
    
Modul uvozimo u memoriju računara kod koristeći komandu `import`. Modul `math` koji sadrži matematičke funkcije uvozimo na sledeći način:

````python
import math
````

Pokretanjem ove komande naglasili smo programu da želimo da koristimo ovaj modul i interpreter traži `math.py` u listi predefinisanih direktorijuma (trenutni direktorijum, i par sistemskih).

Ako se modul ne nalazi u listi predefisnisanih direktorijuma, onda prvo morate da ga instalirate (sačuvate u memoriji na disku) modul kako biste pokrenuli

Pomoću `help(math)` možemo da proverimo koje sve funkcije su nam dostupne u okviru ovog modula

````python
help(math)
````

#### Funkcija `dir()`
    
Ugrađena funkcija koja takođe može da bude korisna kod uvida u funkcije modula. Funkcija vraća sortiranu listu stringova svih funkcija i konstanti koje postoje unutar modula.

````python
dir(math)
````

### Pokretanje funkcija iz modula

Pokretanje funkcija (ali i ostalih delova modula) pozivamo tako što unesemo naziv modula, tačku, a zatim i željenu funkciju. Ako na primer želimo da izračunamo $e^{2.45}$, to možemo učiniti na sledeći način:

````python
math.exp(2.45)
````

### Promenljive modula

Modul može imati svoje konstante. Pozivaju se slično kao i funkcije. Modul `math` ima recimo promenljive kao što su $\pi$, ali i $\infty$.

````python
math.pi
````

````python
math.inf
````

### Skraćivanje poziva modula

Prilikom složenijih operacija biće nezgodno stalno se pozivati na ime modula prilikom poziva funkcija ili promenljivih. To postižemo na sledeći način

```python 
`import package as keyword`
```

````python
import math as m
m.log(11.588346719223392)
````

### Dodatne mogućnosti skraćivanja poziva

Poziv se dodatno može skratiti na sledeći način:

````python
from math import *
math.sin(math.pi/2)
````

Na ovaj način uvoze se sva imena osim onih koja počinju sa donjom crtom (_). Iako na prvi pogled najbolja opcija, savet je da ne koristite ovu mogućnost, jer ona uvodi nepoznati skup imena u interpreter, možda skrivajući neke stvari koje ste već definisali.

Još jedan razlog je što na ovaj način imamo i slabo čitljiv kod, jedna od važnih stvari prilikom rada u bilo kom programu.

Ubacivanje modula u Python nosi sa sobom standardne skraćenice koje se koriste. Tako imamo:

```python
import math as m
import pandas as pd
```

### Ušteda memorijskog prostora

Često se može desiti da vam je modul potreban samo zbog jedne specifične funkcije koju sadrži. Kako ne biste zauzeli veliki memorijski prostor uvlačenjem svih funkcija i promenljivih te metode, možete uraditi sledeće:

````python
from math import exp as e

e(3.45)
````

Naravno, nekada vam treba više od jedne funkcije. I ta opcija je moguća

````python
from math import exp, log
log(exp(3.45))
````

### Čemu zaista služe moduli?

Pogledajte ovo:

````python
print(.1 + .2)
````

Python ne može tačno da predstavi neke decimalne brojeve u binarnom sistemu, pa `0.1 + 0.2` nije tačno.
Da dobiješ tačan decimalni rezultat koristi `Decimal` iz `decimal` modula.

````python
from decimal import Decimal

float(Decimal("0.1") + Decimal("0.2"))
````
