## Ciljevi i ishodi
Ovaj modul pruža uvod u programiranje u Pajtonu za studente ekonomije bez prethodnog iskustva. Fokus je na primeni analitičkih alata za obradu i analizu podataka, simulacije poslovnih procesa i učenje osnovnih optimizacionih metoda relevantnih za odlučivanje u ekonomskom kontekstu.

Nakon završetka modula, studenti će biti sposobni da samostalno pišu Pajton skripte, koriste ključne biblioteke za rad sa podacima, analiziraju realne ekonomske podatke i primenjuju optimizacione i simulacione tehnike u praktičnim poslovnim situacijama.



### Zašto Pajton?

Pajton je trenutno jedan od najpopularnijih programskih jezika u oblasti analize podataka.
Za razliku od jezika R, Python je jezik opšte namene i omogućava široki spektar primena van statistike: razvoj aplikacija, automatizaciju, rad sa bazama podataka, optimizaciju itd.
Python je besplatan (open source) i javno dostupan.
Postoji velika i aktivna zajednica korisnika koja obezbeđuje dokumentaciju, biblioteke i edukativne resurse.
Python je jednostavan za učenje i čitanje, što ga čini pogodnim za početnike.
