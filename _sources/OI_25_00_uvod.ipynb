## Ciljevi i ishodi

<br>

Prvi dva dela ovog udžbenika pružaju uvod u programiranje u Pajtonu za studente ekonomije bez prethodnog iskustva. Fokus je na primeni analitičkih alata za obradu i analizu podataka, simulacije poslovnih procesa i učenje osnovnih optimizacionih metoda relevantnih za odlučivanje u ekonomskom kontekstu. Treći deo udžbenika će studentima ekonomije približiti osnovne koncepte i metode mašinskog učenja, sa posebnim naglaskom na njihovu primenu u analizi ekonomskih i poslovnih podataka. Pored razumevanja principa na kojima se zasnivaju savremeni algoritmi učenja iz podataka, cilj je i da se razvije sposobnost kritičkog sagledavanja mogućnosti i ograničenja ovih metoda u ekonomskim istraživanjima i poslovnoj praksi.

Nakon završetka modula, studenti će biti sposobni da samostalno pišu Pajton skripte, koriste ključne biblioteke za rad sa podacima, analiziraju realne ekonomske podatke i primenjuju optimizacione i simulacione tehnike u praktičnim poslovnim situacijama. Nakon proučavanja trećeg dela ovog udžbenika student će biti u stanju da razume osnovne ideje mašinskog učenja i razlikuje ključne pristupe poput nadgledanog i nenadgledanog učenja. Takođe će moći da primeni osnovne metode mašinskog učenja na ekonomske skupove podataka i da interpretira dobijene rezultate u kontekstu ekonomskih i poslovnih problema.


### Zašto Pajton?

Pajton je trenutno jedan od najpopularnijih programskih jezika u oblasti analize podataka.
Za razliku od jezika R, Python je jezik opšte namene i omogućava široki spektar primena van statistike: razvoj aplikacija, automatizaciju, rad sa bazama podataka, optimizaciju itd.
Python je besplatan (open source) i javno dostupan.
Postoji velika i aktivna zajednica korisnika koja obezbeđuje dokumentaciju, biblioteke i edukativne resurse.
Python je jednostavan za učenje i čitanje, što ga čini pogodnim za početnike.
