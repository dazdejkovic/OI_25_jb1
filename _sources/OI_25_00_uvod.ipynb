U savremenom poslovnom i ekonomskom okruženju, donošenje odluka sve više se oslanja na analizu podataka, kvantitativne modele i algoritamske pristupe. Organizacije se suočavaju sa kompleksnim problemima koji uključuju ograničene resurse, neizvesnost i dinamične promene, zbog čega intuitivno odlučivanje često nije dovoljno. U takvim uslovima, sposobnost da se problemi formalizuju, analiziraju i reše uz pomoć računarskih alata postaje ključna kompetencija.

Knjiga pred vama predstavlja doprinos autora da znanja iz oblasti analitike, operacionih istraživanja i mašinskog učenja, pojmova koji su danas sveprisutni u poslovanju, približe studentima i omoguće prvi korak u njihovom savladavanju. Napravljena je primarno kao udžbenik za dva predmeta na osnovnim studijama: Operaciona istraživanja i Mašinsko učenje. Osnovna ideja je da ova dva predmeta predstavljaju jednu celinu koja uvodi studente u svet rešavanja poslovnih problema pomoću savremenih analitičkih alata.

Poseban naglasak stavljen je na povezivanje teorijskih koncepata sa praktičnom implementacijom, čime se omogućava prelazak od apstraktnih modela ka konkretnim rešenjima. Cilj knjige nije samo da prikaže kako se problemi rešavaju, već i da radi na razvoju kompetencija povezanih sa razumevanjem rezultata. Šta je to što kao budući menadžer ili analitičar treba prepoznati u podacima i koji su ključni uvidi za donošenje odluka i izveštavanje. Razvoj veštačke inteligencije dodatno je unapredio ove pristupe i učinio ih dostupnijim i primenljivijim nego ikada ranije. Sposobnost kritičkog razumevanja problema i kreiranje uvida predstavljaju kompetencije budućnosti. 

Operaciona istraživanja predstavljaju disciplinu koja pruža metodološki okvir za rešavanje ovakvih problema, kroz razvoj matematičkih modela i optimizacionih tehnika. Istovremeno, savremeni razvoj tehnologije omogućio je da se ovi modeli implementiraju i primenjuju uz pomoć programskih jezika i specijalizovanih biblioteka, čime se značajno proširuje njihov praktični domet. Pajton se u tom kontekstu izdvojio kao jedan od najvažnijih alata za analizu podataka, optimizaciju i razvoj modela.

Mašinsko učenje predstavlja prirodno proširenje ovakvog analitičkog pristupa, sa fokusom na otkrivanje obrazaca u podacima i izgradnju modela koji mogu da predviđaju buduće događaje ili ponašanja. Za razliku od klasičnih optimizacionih modela, koji polaze od jasno definisane strukture problema, metode mašinskog učenja uče iz podataka i omogućavaju donošenje odluka u uslovima visoke neizvesnosti i kompleksnosti. Njihova primena danas obuhvata širok spektar oblasti, od procene kreditnog rizika (čemu je posvećena i jedna studija slučaja), preko segmentacije korisnika, do preporuka proizvoda i automatizovanog odlučivanja.

Knjiga je organizovana u tri međusobno povezane celine. U prvom delu obrađuju se osnovni koncepti programiranja u Pajtonu, uključujući strukture podataka, funkcije, rad sa datotekama, upravljanje greškama i dajemo osnove objektno orijentisanog programiranja. Ovaj deo postavlja neophodne temelje za dalji rad i omogućava čitaocu da stekne sigurnost u pisanju, ali pre svega u razumevanju koda.

Drugi deo knjige fokusiran je na primenu Pajtona u operacionim istraživanjima i analizi podataka. Obrađuju se ključne biblioteke kao što su NumPy, pandas i Matplotlib, kao i alati za rešavanje optimizacionih problema u okviru biblioteke SciPy. Posebna pažnja posvećena je Monte Karlo simulacijama kao pristupu modeliranju neizvesnosti i analizi kompleksnih sistema, jer ovu temu smatramo izuzetno važnom, a nedovoljno zastupljenom u silabusima predmeta koje studenti Ekonomskog fakulteta u Beogradu do tada slušaju.

Treći deo knjige uvodi čitaoca u oblast mašinskog učenja, sa fokusom na nadgledano učenje i osnovne algoritme klasifikacije, uključujući naivne Bajesove klasifikatore, k-najbližih suseda, stabla odlučivanja i ansambl metode. Ovaj deo proširuje analitički okvir knjige ka prediktivnom modeliranju i savremenim pristupima obradi podataka.

Kroz ovako strukturisan pristup, knjiga omogućava postepeni razvoj znanja. Krećemo od osnovnih programerskih veština, preko modeliranja i optimizacije, do primene metoda mašinskog učenja. Na taj način čitalac stiče kompetencije neophodne za rešavanje realnih problema i donošenje odluka zasnovanih na podacima.
