### Indeksiranje stringova

Svakom karakteru stringa pristupamo vrlo jednostavno. Stringovi su _indeksirani_, tako što prvi karakter ima indeks 0.

````python
s = 'Jedan kratak niz'
s[1] # karakter na poziciji 2
````

Indeksi mogu biti i negativni brojevi, kada brojanje počinje sa desne strane

````python
s[-1]  #Poslednji karakter
````

````python
s[-6] #šesti karakter sdesna
````

Naravno, ako pokušamo da pristupimo nepostojećem indeksu, Python interpreter će prijaviti grešku

````python
s[17]
````

### Isecanje (engl. slicing)

Dok se indeksiranje koristi kako bi izdvojili pojedinačni karakter, isecanje (često se kaže i slajsing) nam omogućava da dobijemo podstring. Izuzetno važan način koji ćemo takođe nadalje koristiti često.

````python
word = "Python"
word[0:2] # karakteri od pozicije 0 (uključujući) do pozicije 2 (isključujući)
````

````python
word[2:5] # karakteri od pozicije 2 (uključujući) do pozicije 5 (isključujući)
````

````python
word[2:-2]
````

Ne moramo navoditi obe vrednosti unutar uglastih zagrada. U tom slučaju sugerišemo da se krene od starta ili da se ide do kraja stringa:

````python
word[:3] #Kreni od starta do karaktera sa indeksom 3 (isključujući njega)
# word[:3] je isto sto i word[0:3]
````

````python
word[3:] # Počevši od karaktera sa indeksom 3 ispiši mi string do kraja
````

Vreme je za kviz

````python
#Python
word[-2:]

# a - Pytho
# b - on
# c - ohtyP
````

````python
#Python
word[-2:-4]

# a - th
# b - ''
# c - Error
````

Jedan od zanimljivih načina da dobijete ceo string, može biti sledeći:

````python
# word = Python
word[:]

# a - Python
# b - Error
# c - :
# d - ""
````

**Obratite pažnju** &#10071; &#10071; &#10071; 

Primena slajsinga van opsega neće dovesti do greške.

</div>

````python
print(word[4:42])

word[42:]
````

Međutim, ovo nije sve. Formalno, isecanje ima tri argumenta 

- `word[::]` neće javiti grešku i nema razlike između ovog ispisa i `word[:]`

Šta može biti treći argument?

- Dužina koraka. Treći argument koristimo ako ne želimo da nam se redom ispisuje ceo string, već na primer, svaki drugi član tog stringa

Formalno u pseudojeziku slajsing definišemo na sledeći način:

- \<ime_stringa\>[(start):(stop):(step)]

````python
# Python
word[::2]
````

````python
s = "Jedan"
s[0::2]

# a - Jed
# b - Je
# c - Jdn
# d - ea
````

Ovo je verovatno i prva naprednija stvar koju smo spomenuli. Igrajte se sa ovim opcijama da dobro razumete kako funkcionišu.

````python
# Test pitanje
# Šta će biti izlaz sledećeg koda?
word = "Python"
word[1::2]

# a - yhn
# b - y
# c - Py
````

````python
# Test pitanje 2
# Šta će biti izlaz sledećeg koda?
# word = "Python"
word[::-1]
````

### Operacije sa stringovima

Operatori + i * mogu se koristiti i kod stringova, sa malo drugačijom namenom:

- `+` - spaja dva stringa u jedan
- `*` - ponavlja vrednost stringa određeni broj puta
- `+=` takođe je i dalje aktivna

Razmislite šta će biti vrednost stringa `s1` kreiranog sledećom linijom koda:

```python
s1 = 3 * 'py' + 'thon'
```

````python
s1 = 3 * "py" + 'thon'
s2 = 'py' + 'thon' * 2 # Primetite kako prioritet operacija i dalje postoji
print(s1)
print(s2)
````

Spajanje stringova neće kreirati razmak između karaktera. Razmake možemo uvesti kao zaseban string:

````python
s2 = 'py' + ' ' + 'py' + " " + 'thon'
print(s2)
````

### Nepromenljivost stringova

Jedna od važnih osobina stringa je njegova nepromenljivost (eng. *immutable object*). Dodeljivanje nove vrednosti nekom elementu stringa nije dozvoljeno.

</div>

````python
word[0] = 'J'
````

````python
word[2:] = 'py'
````

Svaki drugačiji string moramo iznova kreirati (bilo unutar neke promenljive ili ne)

````python
'J' + word[1:]
````
