# Klase


Zamislite da radite u velikoj fabrici igračaka. Svakog dana treba da se proizvede na hiljade istih modela – recimo, plišanih medveda. Svaki medved treba da ima ime, boju krzna, broj godina (ako baš hoćete da bude realističan), kao i nekoliko tipičnih ponašanja: može da se predstavi, da kaže koliko ima godina i, naravno, da mahne šapom.

Sad zamislite kako bi izgledalo kada bi svaki radnik pravio medvede po svom osećaju. Jedan bi zaboravio da doda boju, drugi bi pogrešno napisao ime, treći bi pustio medveda bez šape… Totalni haos.

Da bi se izbegla takva zbrka, fabrika uvodi nacrt proizvodnje – dokument koji tačno propisuje kako izgleda svaki medved i šta svaki medved ume. Taj nacrt ne proizvodi nijednog medveda direktno; on samo opisuje kakav bi svaki medved trebalo da bude. Tek kad se nacrt upotrebi u proizvodnji, nastane stvarni medved – potpuno funkcionalna igračka sa svim predviđenim osobinama.

U programiranju, taj nacrt zovemo klasa, a medvedi napravljeni na osnovu nacrta su objekti. Definisanjem sopstvenih klasa, programer uvodi nove tipove podataka koji precizno modeluju specifične pojave ili entitete iz realnog sveta.

Klasa definiše:
- koje podatke svaki objekat ima (npr. ime, boja, godine),
- koje funkcije svaki objekat ume da izvrši (npr. predstavi_se(), mahni()).

Objekat je konkretan primerak — pravi, jedinstveni „medved“ sa stvarnim vrednostima tih podataka. Na primer, možemo imati klasu Medved, a njene instance mogu biti:
- Bobi, braon, star 3 godine,
- Luna, bela, stara 1 godinu,
- Grizli, tamno braon, star 5 godina.

Svi potiču iz iste klase, ali svaki je svoj, različit i nezavisan.

I baš kao u fabrici igračaka, klase nam omogućavaju da u Pajtonu pravimo sopstvene „modela“ podataka: čiste, standardizovane i lako ponovljive. Od tog trenutka, kreiranje novih objekata postaje jednostavno kao pokretanje proizvodne trake.

## Definicija klase u Pajtonu.

Za kreiranje klase u Pajtonu koristimo ključnu reč `class`. Na primer:

```Python
class ImeKlase:
    # definicija klase
```
kreira klasu sa imenom `ImeKlase`

Sledećim kodom kreirana je jedna klasa:

````python
class Zaposleni:
    ime = ""
    plata = 0
````

Gde je,
- Zaposleni - ime klase
- ime/plata - promenljive u klasi sa podrazumevanim vrednostima `""` i `0` tim redom

Promenljive unutar klase nazivamo atributima.

### Pajton objekti

Objekat je instanca klase. Na primer, pretpostavimo da je `Zaposleni` klasa, tada možemo da kreiramo dva objekta poput `z1`, `z2`, itd iz te klase.

Sledi sintaksa za kreiranje objekata
```Python
objectIme = Imeklase()

Na primer, pretpostavimo da je `Zaposleni` klasa, tada možemo da kreiramo dva objekta poput `z1`, `z2`, itd iz te klase:

````python
# Kreiranje klase
class Zaposleni:
    ime = ""
    plata = 0
    
# kreiranja objekta iz klase Zaposleni
z1 = Zaposleni()
````

Ovde, `z1` je objekat klase. Sada možemo da upotrebimo ovaj objekat da pristupimo atribitima klase. Koristimo notaciju sa `.` da pristupimo objektima klase:

````python
# Modifikovanje vrednosti atributa
z1.ime  = 'Marković Marko'
z1.plata= 110000
````

Sada možemo da ih koristimo u kod kao i sve druge objekte:

````python
# Upotreba novih atributa
print(f"{z1.ime} ima platu od {z1.plata} dinara")
````

Možemo da kreiramo više objekata iz iste klase

````python
# kreiranje drugog objekta u klasi zaposleni
z2 = Zaposleni()

# pristup atributima objekta z2
z2.ime = 'Jovanović Jelena'
z2.plata = 120000


# pristup atributima objekata z1 i z2
print(f"{z1.ime} ima platu od {z1.plata} dinara")
print(f"{z2.ime} ima platu od {z2.plata} dinara")
````

### Metodi klase.

Unutar klase možemo definisati funkcije. Pajton funkcije unutar klase se nazivaju metodi. Pogledajmo primer,

````python
# Kreiranje klase
class Soba:
    dužina = 0.0
    širina = 0.0
    
# metod za izračunavanje površine
    def površina(self):
        print("Površina sobe =", self.dužina * self.širina)

# kreiranje objekta klase Soba
kuhinja = Soba()

# dodeljivanje vrednosti svim atributima
kuhinja.dužina = 3.4
kuhinja.širina = 2.8

# pristup metodu unutar klase
kuhinja.površina()
````

U prethodnom primeru, kreirali smo klasu sa imenom `Soba`  sa

- **Atributima**: `dužina` i `širina`
- **Metodom**: `površina`  

Kreirali smo objekat sa imenom `kuhinja` iz klase `Soba`. Zatim smo iskoristili taj objekat da dodelimo vrednosti atributima: `dužina` i `širina`. Takođe, upotrebili smo objekat da pozovemo metod iz klase `kuhinja.površina()`, gde smo iskoristili notaciju sa `.` za pozivanje metoda. 

Primetite neobično ime argumenta `self` u metodu. To ime nije rezervisana reč, ali je uobičajeno i koristi se da ukaže na sam objekat!

### Konstruktor  `__init__`

U prethodnim primerima dodeljivali smo podrazumevane vrednosti za atribute klase. Međutim, vrednosti možemo da inicijalizujemo primenom specijalnog metoda `__init__`. Na primer,

````python
class Zaposleni:

    def __init__(self, ime = '', plata = 0):
        self.ime = ime
        self.plata = plata
````

Ovde je `__init__()` specijalna funkcija (funkcija konstruktor) koja se poziva kadgod se novi objekat iz klase instancira. Konstruktor iznad instancira vrednosti atributa `ime` i `plata`. Upotrebili smo `self.ime` i `self.plata` da referišemo na atribute `ime` i `plata` objekata iz klase `Zaposleni`.

Ako koristimo konstruktor da inicijalizujemo vrednosti unutar klase, tada je neophodno prolediti odgovarajuću vrednost za vreme kreiranja objekta iz klase:

````python
z3 = Zaposleni('Ilić Jovan', 130000)
z3.ime
````

Ovde, `'Ilić Jovan'` se prosleđuje parametru `ime`, a 130000 parametru `plata` konstruktor funkciji `__init__()`. 

```{admonition} Važno.
:class: important

Pajton stil je da imena specijalnih metoda i atributa počinju i završavaju sa jednom ili dve donje crte. Kasnije ćemo se upoznati sa još nekim specijalnim metodama i atributima kod klasa.
```

### Atributi klase i atributi instanci

Atributi, prikazani u prethodnom primeru su u stvari atributi instanci, što znači da oni pripadaju samo specifičnoj instanci. Kada ih koristimo, neophodno da upotrebimo `self.atribut` unutar klase. 

Atributi klase se dodelju svim instancama kreiranim u toj klasi. Njima možemo pristupiti na dva načina: `ImeKlase.atribut` ili `ImeObjekta.atribut`.

Ako ponovo pogledamo prvi primer gde smo definisali prvu klasu primetićemo da smo tamo definisali atribute klase. Instance iz te klase preuzele su te atribute, pa smo onda morali da ih modifikujemo u okviru svake instance. Očigledno, to je malo komplikovano, zato postoji `__init__` metod koji postupak incijalizacije rešava vrlo efikasno. 

Atributi klase u principu imaju drugu namenu. Pogledajmo na jednom primeru kako možemo da definišemo i upotrebimo atribute klase.

```{admonition} Primer 2.
:class: tip 

Definišite klasu `Zaposleni` sa atributima `ime`, `plata`, `ime_kompanije = "MTS"` i `broj_zaposlenih` i metodima koji će dati osnovne informacije o zaposlenom, i ukupan broj zaposlenih. Testirajte klasu.  

````python
class Zaposleni:
    # atributi (promenljive) klase
    ime_kompanije = 'MTS'
    broj_zaposlenih = 0

    # Konstruktor za incijalizaciju objekta
    def __init__(self, ime, plata):
        # atributi (promenjive) instance
        self.ime = ime
        self.plata = plata
        Zaposleni.broj_zaposlenih += 1

    # metod instance
    def info(self):
        print(f"Zaposleni {self.ime} radi u kompaniji {self.ime_kompanije} i ima platu {self.plata}.")
        
    # metod klase
    def ukupno(self):
        print(f"U kompaniji radi {self.broj_zaposlenih} radnika")  
````

````python
#Kreiranje prvog objekta
zap1 = Zaposleni("Marko", 120000)
zap1.info()
zap1.ukupno()
````

````python
#Kreiranje drugog objekta
zap2 = Zaposleni("Ivan", 120000)
zap2.info()
zap2.ukupno()
````

### Specijalne metode

Kada ukucate `x=2+3`, kako `integer` objekti sa vrednostima 2 i 3 znaju kako se implementira `+`? Takođe kako zna da upotrebi `=` da se dođe do rezultata? Do ovih znakova operacija možete doći koristeći  Pythonove specijalne metode.

Imena ovih metoda počinju i završavaju se sa duplm donjim crtama (`__`). Jedan smo već videli: `__init__`, pokreće novostvoreni objekat iz definicije njegove klase i sve argumente koji su prosleđeni unutra.

Pretpostavimo da imamo jednostavnu klasu `Reč` i da želimo da proverimo da li su dve reči jednake ignorišući veličinu slova.

````python
class Reč():
    def __init__(self,text):
        self.text = text
    def __eq__(self,reč2):
        return self.text.lower()==reč2.text.lower()
````

Da vidimo da li radi:

````python
r1 = Reč('Ah')
r2 = Reč('ah')
r3 = Reč('Aha')
````

````python
r1==r2
````

````python
r1==r3
````

Pored `__init__()`, najviše se koriste metodi `__str__()` i `__repr__()`. Radi se o tome kako se ispisuje objekat prilikom štampe ili posle izvršavanja neke naredbe. Ukoliko ih ne definišete, Pajton će vam automatski dati string verziju vašeg objekta:

````python
r1
````

Hajde da dodamo i `__str__()` i `__repr__()` metode u klasu `Reč` kako bi je ulepšali:

````python
class Reč():
    
    def __init__(self,text):
        self.text = text
        
    def __eq__(self,reč2):
        return self.text.lower()==reč2.text.lower()
    
    def __str__(self):
        return self.text
    
    def __repr__(self):
        return 'Reč(' + self.text +')'    
````

````python
r1=Reč("Студент")
r1  #koristi __repr__
````

````python
print(r1) #koristi __str__
````

## Nasleđivanje

Nasleđivanje nam omogućava da definišemo klasu koja će naslediti sve metode i atribute iz neke druge klase. Konvencionalno, za novu klasu tog tipa koristimo termin *klasa dete* (child class), dok klasu koju ona nasleđuje zovemo *klasa roditelj* (parent class) ili *super klasa*. Nasleđivanje izgrađuje relaciju između klase dece i klase roditelja. Obično, klasa roditelja je opšteg tipa dok je klasa dece specifičnog tipa.

:::{admonition} Apstraktna sintaksa nasleđivanja
:class: important

```Python
# definicija superklase
class super_klasa:
    # definicija atributa i metoda

# nasleđivanje
class pod_klasa(super_klasa):
    # atributi i metodi super_klase
    # atributi i metod pod_klase
```

Ilustrujmo nasleđivanje primerom:

````python
class Animal:

    def __init__(self, ime):
        self.ime = ime
    
    def eat(self):
        print("Mogu da se hranim.")

# Nasleđivanje iz klase Animal
class Pas(Animal):

    # Novi metod u podklasi
    def display(self):
        # pristup atributu ime iz super klase primenom self
        print(f"Moje ime je  {self.ime}.")
````

````python
# Kreiranje objekta u podklasi
labrador = Pas("Roki")  # Koristi metod __init__ iz super klase
labrador.eat()          # Koristi metod eat super klase
````

````python
# poziv metoda podklase
labrador.display()
````

### Mutacija (premošćavanje, nadvlađivanje metoda)

U prethodnom primeru, videli smo da objekat u podklasi pristupa metodi iz super klase. Međutim, šta ako isti metod postoji u superklasi i podklasi? Tada, metod podklase nadjačava metod superklase.

````python
class Animal:

    # atributi i metodi klase roditelja     
    def __init__(self, ime):
        self.ime = ime
    def eat(self):
        print("Mogu da se hranim.")

# Nasleđivanje iz klase Animal
class Pas(Animal):

    # novi metod u podklasi
    def display(self):
        # pristup atributu ime iz super klase primenom self
        print(f"Moje ime je {self.ime}.")
        
     # nadjačavanje metoda superklase   
    def eat(self):
        print("Volim da jedem koske.")
````

````python
# Kreiranje objekta u podklasi
labrador = Pas("Roki")  
labrador.eat()
````

### Metod `super` u nasleđivanju

Malopre smo videli da metod iz podklase nadjačava metod iz superklase. Međutim, ako nam treba metod sa istim imenom iz superklase, tada koristimo metod `super()`. Na primer,

````python
class Animal:

    # atributi i metodi klase roditelja     
    def __init__(self, ime):
        self.ime = ime
    def eat(self):
        print("Mogu da se hranim.")

# Nasleđivanje iz klase Animal
class Pas(Animal):

    # novi metod u podklasi
    def display(self):
        # pristup atributu ime iz super klase primenom self
        print(f"Moje ime je {self.ime}.")
        
        # nadjačavanje metoda superklase   
    def eat(self):
        print("Volim da jedem koske.")
        
        # poziv metoda iz super klase
        super().eat()
````

````python
# Kreiranje objekta u podklasi
labrador = Pas("Roki")  
labrador.eat()
````

*Napomena*: Vrlo često se premošćava `__init__` metod.

## Enkapsulacija

*Enkapsulacija* je jedan od bazičnih koncepata u OOP. Ona opisuje ideju restrikcije pristupa metodima i atributima u klasi. Enkapsulacija sakriva kompleksne detalje od korisnika i prevencija je promene podataka nekim incidentom. U Pajtonu to se postiže upotrebom privatnih metoda ili atributa koristeći donju crtu kao prefiks, odnosno, jedna `_`  ili dve `__`, kako je to pokazano u sledećem primeru.

````python
class Sensor():
    def __init__(self, ime, lokacija):
        self.ime = ime
        self._lokacija = lokacija
        self.__verzija = '1.0'
    
    # a getter funkcija
    def get_version(self):
        print(f'Verzija senzora je {self.__verzija}')
    
    # a setter funkcija
    def set_version(self, verzija):
        self.__verzija = verzija
````

````python
sensor1 = Sensor('Acc', 'Beograd')
print(sensor1.ime)
print(sensor1._lokacija)
print(sensor1.__verzija)
````

Prethodni primer pokazuje kako radi enkapsulacija: 
- Sa jednom donjom crtom, mi definišemo privatnu promenljivu kojoj se ne može direktno pristupiti. Primetite da je ovo konvencija i ništa vas ne sprečava da joj konkretno pristupite. 
- Sa dve  donje crte, vidimo da se atributu `__verzija__` ne može direktno ni pristupiti ni modifikovati.
  
Kako bi pristupili atributima ćije ime počinje sa dve donje crte, neophodno je da upotrebimo takozvane `getter` i `setter` funkcije kako bi im interno pristupili. 

````python
sensor1.get_version()
````

````python
sensor1.set_version('2.0')
sensor1.get_version()
````

Dodajmo predhodnoj klasi dve specijalne metode koje će na prirodan način prikazivati objekte, ispitivati njihovu jednakost i reprezentovati njihovu prirodu.

Testirati novu klasu.

**Rešenje:**

````python
class Tacka2D:
    
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __str__(self):                         # String reprezentacija objekta
        return (f'({self.x}, {self.y})')
        
    def __eq__(self, drugi):                   # Definicija kad su dva objekta jednaka
        if self.x==drugi.x and self.y==drugi.y:
            return True
        else:
            return False
    
    def __repr__(self):                        # Prikazuje način nastanka objekta
        return f'Tacka2D({self.x},{self.y})'
````

````python

A = Tacka2D (2,2)
B = Tacka2D (2,3)
C = Tacka2D (2,2)

print(A) # Primenjuje str
print(type(A))

print(A==B)  # Primenjuje proveru jednakosti
print (A==C)
````
