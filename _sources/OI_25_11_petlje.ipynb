# Petlje
<br>

Petlja u Pajtonu je konstrukcija koja omogućava да се neki blok koda ponavlja više puta, sve dok postoji neki uslov (kod while) ili dok ima elemenata koje treba obići (kod for). Drugim rečima: petlja služi da automatizuje ponavljanje — da ne pišeš isti kod deset, sto ili hiljadu puta, već da Pajtonn to radi umesto tebe.

## For petlje

**For petlja** je skup instrukcija koje se ponavljaju, ili iteriraju, za svaku vrednost u iterabilnom objektu. Stringovi, liste i torke su primeri struktura podataka koje zovemo *iterabilni objekti*: to su uređeni nizovi predmeta (karakteri u slučaju stringova, ili proizvoljni objekati u slučaju listi i torki).

:::{admonition} Apstraktna sintaksa for petlje
:class: important

~~~python
for promenljiva_petlje in iterabilni_objekat:
    blok koda
~~~
:::

- For petlja dodeljuje promenljivoj petlje prvi element iterabilnog objekta.
- Izvršava sve naredbe u bloku koda
- Ako postoji sledeći element iterabilnog objekta, tada dodeljuje promenljivoj petlje sledeći element iterabilnog objekta i ponovo izračunava blok koda
- Nastavlja sve dok ne postoji više elemenata u iterabilnom objektu za dodeljivanje.
    


### Funkcija `range`

Ako treba ponavljati neku radnju kroz niz (iterirati), koristimo ugrađenu funkciju `range`. Ona generiše aritmetičku progresiju. Pozivamo je na sledeći način:

:::python
range(start = 0, stop, step = 1)
:::

 - `start` je prvi element progresije. 
 - `stop` definiše do kog broja ide progresija, ali nije član progresije.
 - `step` je korak progresije, odnosno, razlika  uzastopnih članova.

Na mnogo načina objekat koji vraća `range()` ponaša se kao da je lista, ali on to ipak nije. To je iterabilni objekat koji vraća uzastopne elemente zahtevanog niza kada iteriramo preko njega, ali on realno ne pravi listu, čuvajući na taj način memorijski prostor.   


:::{admonition} Primer 1.
:class: tip

Izračunajmo zbir svih celih brojeva od 1 do 3.
:::

````python
suma = 0
for i in range(1,4):
    suma = suma + i
print(suma)
````

Šta se dešava?
1. Prvo, funkcija `range(1,4)` generiše niz celih brojeva koji počinju od 1 i završavaju se sa 3. 
2. Promenljivoj `suma` se dodeljuje vrednost 0
3. Promenljivoj `i` se dodeljuje vrednost 1.
4. Promenljivoj `suma` se dodeljuje vrednost `suma + i` (0+1=1).
5. Promenljivoj `i` se dodeljuje vrednost 2.
6. Promenljivoj `suma` se dodeljuje vrednost `sum + i` (1+2=3).
7. Promenljivoj `i` se dodeljuje vrednost 3.
8. Promenljivoj `suma` se dodeljuje vrednost `suma + i` (3+3=6).
9. Nema više vrednosti koje bi se dodelile promenljivoj `i`, prekida se for petlja.

## Petlje i rečnici

Sada kada znamo da prolazimo iterabilno kroz strukturu podataka dobijamo puno novih mogućnosti koje su posebno interesantne za rečnike. Podsetite se rečnika i istražite i sami šta sve mogu raditi petlje u rečnicima. Proći ćemo metode `.values`, `.keys` i `.items` koje preporučujemo da sami dalje istražite i bolje razumete.

:::{admonition} Primer 5.
:class: tip

Izračunati zbir svih vrednosti u rečniku `recnik_a = {'A': 1, 'B':2, 'C': 5}`.
:::

````python
recnik_a = {'A': 1, 'B':2, 'C': 5}
suma = 0
for x in recnik_a.values():
    suma += x
print(suma)   
````

U primeru, koristimo prvo metod `values`da dobijemo sve vrednosti. Zatim, iteriramo preko niza vrednosti.

Alternativno, možemo zadatak rešiti pristupajući nizu ključeva

````python
recnik_a = {'A': 1, 'B':2, 'C': 5}
suma = 0
for x in recnik_a.keys():
    suma = suma + recnik_a[x]
print(suma)   
````

Šta mislite da će biti rezultat sledeće linije koda?

````python
type(recnik_a.keys())
````

Pretpostavimo da sada želimo da štampamo sve ključeve i vrednosti iz prethodnog rečnika. Koristimo metod `items` u rečniku. On vraća objekat čiji je sadržaj torka parova ključeva i vrednosti. Možemo ih dodeliti simultano dvema promenljivim (dodeljivanje torki); pogledaj kod koji sledi ...

````python
recnik_a = {'A': 1, 'B':2, 'C': 5}
for key, value in recnik_a.items():
    print(key, value) # koristimo key, value kao promenljive, tu naravno može ići bilo šta
````

Primetite da smo dodelili vrednosti dvema promenljivima u petlji u isto vreme. Postoje i drugi slučajevi gde možemo simultano dodeliti poslove. Na primer, ako imamo dve liste iste dužine i želimo petlju kroz njih simultano, tada koristimo `zip` funkciju. Pogledajmo primer koji sledi.

## Funkcija zip

Funkcija `zip` agregira elemente iz dva iterabilna objekta i vraća iterator torki, gde i-ta torka sadrži i-ti element svakog 
iterabilnog objekta.   


````python
a = ["A", "B", "C"]
b = (1,2,3)
for i,j in zip (a,b):
    print (i,j)
````

## Funkcija enumerate

S obzirom da range objekat može da se iskoristi za kreiranje niza celih brojeva, primamljivo je da se oni iskoriste kao indeksi za liste ili torke kada iteriramo preko njih u for petlji.    


````python
sisari = ['kengur', 'kljunar', 'delfin']
for i in range(len(sisari)):
    print(i,':', sisari[i])     
````

Ovo radi, naravno, ali je prirodnije izbeći eksplicitnu konstrukciju range objekta (i pozivanje ugrađene len funkcije) primenom funkcije `enumerate`

````python
sisari = ['kengur', 'kljunar', 'delfin']
for i, životinja in enumerate(sisari):
    print(i,':', životinja)   
````

## Komanda break

Ako se izvrši komanda `break` unutar bloka koda `for` petlje momentalno se izlazi iz petlje.


:::{admonition} Primer 6.
:class: tip
Korisnik unosi string. Odrediti da li string ima cifru u svom zapisu. Ispisati prvu cifru koja se pojavljuje, ako postoji. Ako ne postoji cifra u zapisu, ispisati da string nema cifara. Razmislite i prodiskutujmo kako bismo kreirali ovaj kod.
:::

*Ideja*: Primenićemo metod `isdigit` za stringove da proverimo da li je neki karakter cifra.

````python
# Prvo rešenje
string = input("Unesite string:")
ind = 0
for c in string:
    if c.isdigit(): 
        ind = 1
        break
if ind: 
    print("String ima cifru", c) 
else:
    print("String nema cifara") 
````

````python
# Drugo rešenje - elegantnije
string = input("Unesite string:")

for c in string:
    if c.isdigit(): 
        print("Prva cifra stringa je", c)
        break
else:
    print("String nema cifara") 
````

## Komanda continue

Ako se izvrši komanda `continue`, preskače se preostali kod u trenutnoj iteraciji `for` petlje, i nastavlja sa sledećim 
elementom iterabilnog objekta koji je u petlji.
    
U sledećem primeru koristimo naredbu `continue` da funkcija `print` preskoči štampanje broja 2.   
  

````python
for i in range(5):
    if i==2:
        continue
    print(i)
````

## Ugneždene for petlje

Kao i `if` naredbe i `for` petlje mogu biti ugneždene.
    
:::{admonition} Primer 7.
:class: tip

Neka je `a` lista liste brojeva. Upotrebite ugneždene for petlje da
saberete sve elemente te liste. Pokušajte u narednih 5 minuta. Proverite vaš kod na test primeru `a = [[2, -6], [1, 7]]`. 
:::

````python
# Kod 1

sum = 0
a = [[2, -6], [1, 7]]

m = len(a)
n = len(a[0])
for i in range(m):
    for j in range(n):
        sum += a[i][j]
sum
````

Šta se dešava?
1. `sum`, koje reprezentuje ukupnu sumu, postavljeno je na 0.
2. Spoljna for petlja počinje sa postavljanjem promenljive petlje, `i`, na 0.
3. Unutrašnja for petlja počinje sa postavljanjem promenljive petlje, `j`, na 0.
4. `sum` se uvećava za `a[i][j] = a[0][0] = 2`, pa je `sum = 2`.
5. Unutrašnja for petlja postavlja promenljivu petlje, `j`, na 1.
6. `sum` se uvećava za `a[i][j] = a[0][1] = -6`, pa je `sum = -4`.
7. Unutrašnja petlja se prekida.
8. Spoljna for petlja postavlja promenljivu petlje, `i`, na 1.
9. Unutrašnja for petlja počinje sa postavljanjem promenljive petlje, `j`, na 0.
10. `sum` se uvećava za `a[i][j] = a[1][0] = 1`, pa je `sum = -3`.
11. Unutrašnja for petlja postavlja promenljivu petlje, `j`, na 1.
12. `sum` se uvećava za `a[i][j] = a[1][1] = 7`, pa je `sum = 4`.
13. Unutrašnja petlja se prekida.
14. Spoljašnja petlja se prekida, sa `sum = 4`.

```{admonition} QA time
:class: hint

Šta biste zamerili kod ispisanom gore? Kako da to popravimo?
```

Ovaj kod pretpostavlja da je lista `a` dimenzija 2x2, odnosno pretpostavlja da drugi element liste ima isti broj članova. Kod će recimo za listu `a = [[2, -6], [1, 7, 3, 6]]` dati istu vrednost rezultata jer neće uzeti u razmatranje brojeve 3 i 6.

````python
# Kod 2

sum = 0
a = [[2, -6], [1, 7, 3, 6]]

for i in range(len(a)):
    for j in range(len(a[i])):
        sum += a[i][j]
sum
````

Pokušajte da isti kod ispišete koristeći funkciju `enumerate`.

````python
sum = 0
a = [[2, -6], [1, 7, 3, 6]]
for i, j in enumerate(a):
#    print(i,j)
    for k in j:
        sum += k
sum
````

## While petlje

**While petlja** je skup instrukcija koje se ponavljaju sve dokle je odgovarajući logički izraz istinit. Sledi apstraktna sintaksa `while` petlje

:::{admonition} Apstraktna sintaksa `while` petlje
:class: important
~~~python    
while logicki_izraz:
    blok koda
~~~
:::
- Kad Python dođe do bloka while petlje, on prvo proverava da li je logički izraz while petlje istinit ili lažan. 
- Ako je izraz istinit, izvršava se blok koda, a zatim se program vraća na logički izraz na početku while naredbe.
- Ako je lažan, prekida se while petlja i prelazi na sledeću naredbu.   

~~~{admonition} Primer 8.
:class: tip

Kreirajte program koji će računati kvadratni koren broja $a$ pomoću numeričke metode navedene ispod. Razmislimo prvo algoritamski, kako bi kod trebalo da izgleda:

```{math}
x_0=5, x_{n+1}=\dfrac12\left(x_n+\frac{a}{x_n}\right), n=2,3,\ldots
```

Proverićemo naš kod na primeru `a = 289`.
~~~

````python
a = 289
x = 5

y=(x+a/x) / 2

while abs(x-y)>0:    
    y,x = x,(x+a/x)/2

x
````

```{admonition} Primer 9.
:class: tip

Interesantan primer primene while petlje je dat kroz implementaciju Euklidovog algoritma za nalaženje najvećeg zajedničkog delioca dva broja $NZD(a,b)$.
```

````python
a, b = 1071, 462
while b:
    a,b = b, a%b
print(a)
````

Šta se dešava?

1. `a` i `b` su brojevi za koje treba odrediti NZD
2. Počinje while petlja.
3. Kako je `b=462`, odgovara mu logička vrednost `True` pa se izvršava blok koda.
5. Promenljivoj `a` se dodeljuje vrednost promenljive `b` tj 462, a `b` dobija vrednost ostatka pri deljenju 1071 sa 462, tj. 147
6. Kako je `b=147`, odgovara mu logička vrednost `True`, izvršava se ponovo blok koda.
7. Sada `a=147`, a `b` dobija vrednost ostatka pri deljenju 462 sa 147, tj. 21
8. Kako je `b=21`, odgovara mu logička vrednost `True`, izvršava se ponovo blok koda.
9. Promenljiva `a` sada dobija vrednost 21, a `b` dobija vrednost ostatka pri deljenju 147 sa 21, tj. 0
10. Konačno, sad je `b` jednako 0, odgovara mu logička vrednost `True` i while petlja se prekida.
11. Trenutna vrednost`b`, 21, reprezentuje NZD za date brojeve.

### Komanda `else`

Posle `for` i `while` petlji može da sledi `else` blok naredbi, koji se izvršava samo ako se petlje završe "normalno", (odnosno bez intervencije komande `break`). Za `for` petlje, to znači da će se naredbe izvršiti kad petlja dostigne kraj niza preko koga se iterira; Za `while` petlje, one se izvršavaju kad uslov `while` postane `False`.

Posmatrajmo program koji pronalazi prvo pojavljivanje negativnog broja u listi.    


````python
lista = [0,4,5,2,5,10]
for i,a  in enumerate(lista):
    if a < 0:
        print(a, 'se pojavljuje na poziciji', i )
        break
else:
    print('Lista ne sadrži negativne brojeve')
````

<em>"Šta se dešava ako je logički izraz istinit i nikad se ne menja?"</em>

## Sagledavanja (Comprehensions, Jezgrovite `for-if` petlje )

U Pajtonu, postoje i drugi načini da napravimo iteracije; Sagledavanja listi, rečnika i skupova su vrlo popularna zbog svoje elegancije i kratkoće. Sagledavanja dozvoljavaju kreiranje novih nizovnih objekata iz datih nizovnih objekata primenom vrlo kompaktne sintakse.

:::{admonition} Apstraktna sintaksa sagledavajuće petlje
:class: important   
```python
[ f(x) for x in ulazni_niz if uslov ]  
```
:::
gde je: 
- f(x) - formula koju koristimo
- ulazni_niz - iterabilna struktura (niz)
- Deo `if uslov` je opcioni (podrazumevana vrednost je True)

Metod `normalvariate` iz modula `random` može da generiše slučajan broj iz normalne raspodele $\mathcal{N}(\mu,\,\sigma^{2})$. Istražite kako taj metod funkcioniše i formirajte uzorak obima 10 iz raspodele $\mathcal{N}(5,\,1.25^2)$ koji treba sačuvati u jednoj listi. Zatim napišite program čiji će izlaz biti broj elemenata koji su veći od 5 u tom nizu.
:::

````python
# Rešenje:

import random as rnd

Lista = [round(rnd.normalvariate(5, 1.25), 2) for i in range(10)]
print('Dobijena lista: ', Lista)

count = 0

for i in Lista: # i uzima vrednosti elemenata u Listi u svakoj iteraciji (biće ih onoliko koliko imamo elemenata)
    if i > 5:
        count += 1

print(count)
````

Pretpostavimo da na vaš štedni račun na kraju godine, pa sve do kraja 2044. ulažete po 2000$. Koliko će vredeti ovaj iznos 2066. godine, ako je kamatna stopa sve vreme  7%?

Učitajte klasu `date` iz modula `datatime`. Metod `year()` može vam preko svog atributa `year` dati godinu u kojoj se nalazimo.

**Napomena**: koristite petlje, a kao proveru možete iskoristiti formulu za složenu kamatu (nije neophodno)
:::

**Rešenje:**

````python
from datetime import date

godina = date.today().year
 
D = 2000
r = 0.07
suma1 = 0

# Prvo dolazimo u trenutak kraj 2044.
for i in range(2044-godina):
    suma1 += D*((1+r)**i)
    
# Sada ovu vrednost treba dovesti u 2066.
s66 = suma1*((1+r)**22)
print(f"Na kraju 2044. godine kada se završe isplate, njihova ukupna vrednost biće {suma1:.2f}, \
\n a ova vrednost će {2066-godina} godine od danas, tj 2066. godine vredeti {s66:.2f}")
````
