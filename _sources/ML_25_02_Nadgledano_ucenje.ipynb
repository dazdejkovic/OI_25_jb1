## Nadgledano učenje

**Nadgledano učenje** (engl. *supervised learning*) je tip mašinskog učenja u okviru kojeg algoritam uči na skupu podataka u kome su poznati i ulazni podaci i odgovarajući izlazni podaci (oznake).

Podaci za obuku sastoje se od parova **ulaz–izlaz** (npr. karakteristika i oznaka), zbog čega se ovakvi skupovi podataka često nazivaju **označenim skupovima podataka** (*labeled datasets*).

Cilj algoritma nadgledanog učenja jeste da nauči **mapiranje (funkciju)** koja povezuje ulazne podatke sa odgovarajućim izlazima. Model se obučava na skupu podataka u kome su za zadate ulazne vrednosti poznate tačne izlazne vrednosti (jedna ili više), kako bi se uočilo pravilo preslikavanja i model osposobio da precizno predvidi rezultate za nove, prethodno neviđene ulaze.

Napomenimo da se polazi od pretpostavke da između ulaznih i izlaznih vrednosti postoji određena veza. Zadatak algoritma je da se trenira na postojećim podacima kako bi se ta veza identifikovala i iskoristila za predviđanje izlaznih vrednosti na osnovu novih ulaznih podataka.



### Ulazni podaci

Svaka instanca u posmatranom skupu podataka predstavljena je kao **vektor ulaznih karakteristika**, odnosno atributa ili promenljivih (engl. *features*).

Ulazni podaci se tipično označavaju sa $x$ i mogu se matematički zapisati u vektorskoj formi:

$$
x = (x_1, x_2, x_3, \ldots, x_n)
$$

gde svaka komponenta vektora predstavlja jednu karakteristiku.

---

### Izlazni podaci

Izlazne vrednosti se najčešće označavaju simbolom $y$ i predstavljaju **ciljnu promenljivu** (engl. *target variable*).

U većini slučajeva $y$ je jednodimenzionalna promenljiva, ali može biti i višedimenzionalna, u zavisnosti od prirode problema i tipa zadatka nadgledanog učenja.

Postoje i situacije u kojima ni $x$ ni $y$ nisu predstavljeni numeričkim vrednostima, već mogu imati složenije strukture, kao što su:
- sekvence,
- grafovi,
- tekstualni podaci,
- ili drugi oblici koji zahtevaju posebne pristupe obrade i modelovanja.



Tokom procesa treniranja modela mašinskog učenja, u okviru **nadgledanog učenja**, sistemu se obezbeđuju **eksplicitni primeri za učenje**, pri čemu su uz ulazne podatke dostupni i **tačni izlazi**. Ovaj proces je analogan situaciji u kojoj se učeniku prikazuju reprezentativni primeri uz jasno definisana objašnjenja, na osnovu kojih se postepeno formira razumevanje pravila i obrazaca.

### Primer iz analize kreditnog rizika

Zamislite da se algoritam obučava da funkcioniše kao **analitičar rizika**, pri čemu se u potpunosti oslanja na unapred obeležene istorijske podatke. Ovi podaci, poznati kao **direktni primeri obuke**, sastoje se od parova:

- **Ulazni vektor** – karakteristike klijenta (npr. primanja, zaduženja),
- **Izlaz / odgovor** – ishod (da li je kredit *„Otplaćen“* ili *„Nije otplaćen“*).

Model koristi ove *apriori* poznate informacije kako bi pronašao funkciju koja povezuje ulaze sa ispravnim izlazima. Podaci dati za obuku moraju biti **veoma tačni i pouzdani**, jer se mašina isključivo oslanja na ovo istorijsko iskustvo prilikom formiranja pravila.

Od presudne je važnosti da podaci korišćeni za obuku budu **izuzetno precizni i konzistentni**. Ukoliko se modelu prezentuju netačne informacije, on **nema sposobnost da samostalno prepozna**. Ukratko:

- ako su podaci **visokog kvaliteta**, model će donositi **pouzdane odluke** o predviđanju rizika;
- ako su podaci **netačni**, performanse sistema biće direktno narušene.

Ovaj **iterativni proces učenja i ispravljanja grešaka** nastavlja se sve dok algoritam ne dostigne željeni, odnosno zadovoljavajući nivo tačnosti i ukupnih performansi modela.

## Primarne primene nadgledanog učenja

U zavisnosti od tipa izlaza koji želimo da dobijemo, modele nadgledanog učenja delimo na dve glavne kategorije:

### 1. Klasifikacija (Prepoznavanje obrazaca)
Model se obučava da svrstava opservacije u **diskretne kategorije** (klase). Njegov zadatak je da prepozna kojoj grupi određeni podatak pripada.

* **Primer:** Klasifikacija elektronske pošte.
* **Pitanje:** „Da li je ovaj mejl *spam* ili *regularna pošta*?“
* **Izlaz:** Oznaka kategorije (npr. `0` ili `1`, `Spam` ili `Ham`).

### 2. Regresija (Numeričko predviđanje)
Za razliku od klasifikacije, ovde model uči da predviđa **kontinuirane numeričke vrednosti**. Cilj je dobiti precizan broj na osnovu trendova u podacima.

* **Primer:** Procena vrednosti nekretnina.
* **Pitanje:** „Koliko će iznositi tržišna cena stana na osnovu njegove kvadrature i lokacije?“
* **Izlaz:** Konkretan broj (npr. `125.500 €`).

## Klasifikacija u mašinskom učenju

**Klasifikacija** je proces predviđanja diskretne oznake klase iz unapred definisanog skupa mogućih vrednosti. U zavisnosti od broja kategorija, klasifikacione zadatke delimo na dve osnovne grupe:

### 1. Binarna klasifikacija
Ovo je poseban slučaj gde model bira između **tačno dve klase**. Najlakše ju je razumeti kao davanje odgovora na **da/ne** pitanja.

* **Primer:** Detekcija neželjene pošte (Spam filter).
* **Pitanje modela:** „Da li je ova poruka *spam*?“
* **Ishodi:** * `Da` (Spam / Pozitivna klasa)
    * `Ne` (Legitimna pošta / Negativna klasa)

### 2. Višeklasna klasifikacija 
Primenjuje se kada je potrebno izvršiti razvrstavanje između **više od dve klase**. Ovde model bira jednu oznaku iz šireg spektra mogućnosti (npr. prepoznavanje rukom pisanih cifara od 0 do 9).

---

| Tip klasifikacije | Broj klasa | Primer pitanja |
| :--- | :--- | :--- |
| **Binarna** | Tačno 2 | Da li je transakcija regularna ili sumnjiva? |
| **Višeklasna** | > 2 | U koju kategoriju kreditnog rizika spada klijent? |

> **Napomena:** Kod binarne klasifikacije često jednu klasu nazivamo "pozitivnom" (ona koju želimo da detektujemo), a drugu "negativnom", bez obzira na njihovu semantičku vrednost. Pozitivna klasa je obično ona koja predstavlja prisustvo određenog stanja, događaja ili anomalije koju model treba da izoluje. Na primer, u zadacima detekcije neželjene pošte, pozitivna klasa se često odnosi na spam poruke. Izbor pozitivne klase nije univerzalan, već predstavlja subjektivnu odluku koja zavisi od oblasti primene.

## Regresioni zadaci (Predviđanje numeričkih vrednosti)

Kod **regresije**, cilj je predvideti **kontinuiranu numeričku vrednost**. Glavna karakteristika regresije je da predviđena vrednost može biti bilo koji broj unutar određenog opsega.

---

### Praktični primeri regresije:

1. **Predviđanje kretanja cena akcija**
U finansijama se regresija koristi za procenu budućih vrednosti akcija analizirajući istorijskih podataka i tržišne indikatore.
   * **Ulaz (Karakteristike):** indeksi volatilnosti, obim trgovanja (volume), kamatne stope, istorijske cene akcija, itd.
   * **Izlaz:** Cene akcije u određenom vremenskom trenutku u budućnosti.

2. **Optimizacija vremena isporuke:**
   * **Ulaz (Karakteristike):** razdaljina u kilometrima, broj mesta za pretovar, trenutni vremenski uslovi i mogućnosti transporta, itd.
   * **Izlaz:** Vreme isporuke izraženo u vremenskoj jedinici.

---

Najjednostavniji kriterijum za razlikovanje klasifikacije od regresije jeste provera **kontinuiteta izlazne promenljive**. Suština je u prirodi vrednosti koju model predviđa: ukoliko između mogućih ishoda postoji **kontinualna veza**, takav problem klasifikujemo kao regresioni zadatak. To znači da izlaz nije sveden na izolovane kategorije, već se kreće unutar neprekidnog opsega gde svaka međuvrednost ima svoje matematičko i logičko značenje. Nasuprot tome, kod klasifikacije tragamo za diskretnim granicama koje razdvajaju objekte u jasno definisane grupe.

## Generalizacija, prekomerno prilagođavanje i nedovoljno prilagođavanje

Glavni cilj **nadgledanog učenja** je razvoj modela koji može da donosi tačne zaključke nad novim, neviđenim podacima, tj. da uspešno **generalizuje**.

### Ključni principi:

* **Sposobnost generalizacije:** Model treba da prepozna osnovni obrazac (pattern), a ne šum u podacima.
* **Evaluacija:** Visoka tačnost na trening skupu nije merilo uspeha. Prava vrednost modela dokazuje se isključivo na **test podacima**.
* **Nedovoljno prilagođavanje:** Javlja se kada je model suviše jednostavan da obuhvati ključne zakonitosti.
* **Kompleksnost vs. Jednostavnost:** Složeni modeli su skloni preprilagođavanju (*overfitting*). 
* **Cilj:** Teži se izboru **što jednostavnijeg modela** koji obezbeđuje zadovoljavajuću preciznost i dobru generalizaciju.

---

> Što veću složenost dozvolimo modelu, to će on bolje predviđati na trening podacima.  
> Međutim, ako model postane previše složen, počinje da se preterano prilagođava pojedinačnim primerima iz trening skupa, zbog čega gubi sposobnost dobre **generalizacije** na novim podacima.

U mašinskom učenju, **klasifikacija** je proces gde algoritmi svrstavaju nove podatke na osnovu njihovih **atributa (karakteristika)** u jednu od definisanih **kategorija**, koje se nazivaju **oznake (engl. labels)**. Problemi klasifikacije predstavljaju specifičnu formu **nadgledanog učenja**.  

U procesima klasifikacije, izlazna (**ciljna**) promenljiva $y$ označava pripadnost određenoj **klasi ili kategoriji**. Formalno, možemo pretpostaviti da važi:  

$$
y \in \{1, 2, \dots, C\}
$$

gde $C$ predstavlja **ukupan broj definisanih klasa**.  

Kao i kod regresionih modela, **ulazni podaci (atributi)** mogu biti različitih tipova:  

* **Numerički** (npr. realne vrednosti, intenzitet piksela, fizička merenja)  
* **Kategorički** (npr. pol, tip zemljišta, nivo obrazovanja)  


**Klasifikacija** predstavlja jednu od najznačajnijih i najšire zastupljenih oblasti **mašinskog učenja**, čiji je cilj da, učeći na primerima sa poznatim ishodima, predviđa **kategoriju** ili **oznaku klase** za nove instance. Klasifikacija predstavlja veštinu **prepoznavanja obrazaca u podacima** kako bi se donela informisana odluka o njihovoj pripadnosti određenim kategorijama.

Primena **klasifikacionih algoritama** prožima brojne savremene servise, često neprimetno za korisnike, gde obavljaju zadatke poput:  

* Automatskog filtriranja nepoželjne elektronske pošte  
* Detekcije prevara u finansijskim transakcijama  
* Procene kreditnog rizika  
* Biometrijske identifikacije pri otključavanju pametnih uređaja  
* Automatske kategorizacije i rangiranja sadržaja na digitalnim platformama


**Tipovi problema klasifikacije**

U zavisnosti od broja dostupnih kategorija (oznaka klasa) u skupu podataka, klasifikaciju najčešće delimo na **binarnu** (engl. Binary Classification), **višeklasnu** (engl. Multiclass Classification) i **klasifikacija sa više oznaka** (engl. Multi-label Classification). 

**Binarna klasifikacija** je najjednostavniji oblik klasifikacije gde ciljna promenljiva $y$ može poprimiti samo jednu od dve moguće vrednosti. Najčešće se formalno zapisuje kao $y \in \{0, 1\}$ ili $y \in \{-1, 1\}$.

Primeri:
* Detekcija spama: Da li je e-mail *Spam* ili *Nije spam*?
* Medicinska dijagnostika: Da li je nalaz *Pozitivan* ili *Negativan*?
* Odobravanje kredita: Da li se zahtev *Prihvata* ili *Odbija*?

**Višeklasna klasifikacija** je potrebno podatke svrstati u tri ili više međusobno isključivih kategorija, govorimo o višeklasnoj klasifikaciji. Ovde ciljna promenljiva poprima vrednost iz skupa $y \in \{1, 2, \dots, C\}$, gde je $C > 2$.

Primeri:
* Računarski vid: Prepoznavanje životinja na fotografiji (npr. *Mačka*, *Pas*, *Ptica*, *Medved*).
* Kategorizacija teksta: Automatsko svrstavanje vesti u rubrike (npr. *Politika*, *Sport*, *Ekonomija*, *Zabava*).
* Prepoznavanje rukopisa: Klasifikacija cifara od 0 do 9.

**Klasifikacija sa više oznaka** podrazumeva situaciju u kojoj jedan podatak istovremeno može biti dodeljen većem broju klasa, jer njegove karakteristike odgovaraju različitim kategorijama.

Primeri:
* Kategorizacija vesti: Jedan članak može biti označen kao Politika, Ekonomija i Međunarodni odnosi istovremeno.
* Analiza slika: Fotografija može sadržati više oznaka, npr. Osoba, Priroda i More.
* Preporuka muzike: Jedna pesma može pripadati žanrovima Pop, Elektronska i Dance.

### Prmeri rada klasifikacionih modela

Princip rada klasifikacionih modela se može ilustrovati na primeru iz bankarskog sektora.  
U okviru sistema za upravljanje rizicima, svakodnevno se analizira veliki broj zahteva za odobravanje kredita.  
Za svaki zahtev neophodna je evaluacija kreditne sposobnosti klijenta, odnosno verovatnoću da će preuzete obaveze biti uredno izmirene.

Razmotrićemo dva scenarija:

#### Scenario 1: Binarna odluka (procenjeni rizik)

Klijent podnosi zahtev za stambeni kredit, zadatak modela analizira njegove ključne atribute, kao što su:  

- mesečna primanja  
- kreditna istorija  
- radni staž  

Na osnovu obrazaca identifikovanih u skupu za treniranje, model sprovodi binarnu klasifikaciju.  
Klijent se mapira u jednu od dve diskretne kategorije:  

- **Nizak rizik** – odobravanje zahteva  
- **Visok rizik** – odbijanje zahteva  

Dakle, algoritmi klasifikacije preuzimaju zadatak da razdvoje dobre od loših platiša.

#### Scenario 2: Višeklasna klasifikacija (Kreditni rejting)

U složenim finansijskim sistemima, odlučivanje se retko zasniva na binarnim ishodima;  
umesto toga, klijenti se procenjuju kroz sistem kreditnog rejtinga, čime se problem svodi na višeklasnu klasifikaciju.  

Analizom finansijskih pokazatelja, kao što su profitabilnost, likvidnost i stabilnost prihoda, model razvrstava klijente u unapred definisane klase kreditnog rejtinga:

- **Klasa A** – investicioni rang sa visokim nivoom sigurnosti i stabilnosti  
- **Klasa B** – špekulativni rang sa umerenim rizikom i većom osetljivošću na tržišne promene  
- **Klasa C** – visok rizik, sa povećanom verovatnoćom neizmirenja obaveza ili bankrota

Preciznost ovakvog klasifikacionog modela ima neposredan uticaj na kapitalni bilans finansijskih institucija.  
Na primer, pogrešna alokacija visokorizičnog klijenta u kategoriju investicionog ranga (Klasa A) može uzrokovati ozbiljne finansijske gubitke usled neizmirenja obaveza.  

U nastavku ovog poglavlja ćemo predstaviti ključne metrike za evaluaciju performansi i pouzdanosti ovakvih klasifikatora.


### Algoritmi za klasifikaciju 

Za rešavanje problema klasifikacije postoji širok spektar algoritama mašinskog učenja (machine learning algorithms).  
Među najzastupljenijima su: 

- Logistička regresija (engl. Logistic Regression)  
- K-najbližih suseda (engl. K-Nearest Neighbors, k-NN)  
- Stabla odlučivanja (engl. Decision Trees)  
- Mašine sa potpornim vektorima (engl. Support Vector Machines, SVM)  
- Naivni Bajes (engl. Naive Bayes)  
- Ansambl metode (engl. Ensemble Methods)  
- Slučajne šume (engl. Random Forest)  
- Gradient Boosting (engl. Gradient Boosting Machines, GBM)  
- Veštačke neuronske mreže (engl. Artificial Neural Networks, ANN)  
- XGBoost (engl. Extreme Gradient Boosting)  
- LightGBM (engl. Light Gradient Boosting Machine)  
- Kombinacije ansambla (engl. Stacking / Voting Classifier)


Izbor algoritma zavisi od specifičnosti problema, kao i od strukture, obima i kompleksnosti samih podataka, budući da svaki model ima svoje prednosti i mane.

### Metrike za evaluaciju uspešnosti klasifikacionih modela

Evaluacija modela vrši se na osnovu njegove sposobnosti da **tačno predvidi klasu novih, do tada neviđenih podataka**.  
Kod problema klasifikacije, kvalitet modela se ne može uvek pouzdano proceniti samo ukupnom tačnošću, naročito u slučajevima kada su klase **neuravnotežene**, što je česta pojava u finansijskim podacima.

Zbog toga se koriste mere zasnovane na **matrici konfuzije** (engl. confusion matrix).

---

## 🔁 Matrica konfuzije (engl. Confusion Matrix)


Matrica konfuzije se koristi za procenu performansi klasifikacionog modela mašinskog učenja.  
Ona tabelarno sistematizuje broj tačnih i pogrešnih predviđanja. Na osnovu ovih podataka izračunavaju se ključne metrike za ocenu kvaliteta samog modela. i omogućava analizu uspešnosti klasifikacije.  

Za binarne klasifikacije, matrica konfuzije prikazuje četiri osnovna ishoda:

- **Istinski pozitivan (engl. True Positive, TP)** – Model je predvideo pozitivno, a stvarno je pozitivno.  
- **Istinski negativni (engl. True Negative, TN)** – Model je predvideo negativno, a stvarno je negativno.  
- **Lažno pozitivan (engl. False Positive, FP)** – Model je predvideo pozitivno, a stvarno je negativno (greška tipa I).  
- **Lažno negativan (engl. False Negative, FN)** – Model je predvideo negativno, a stvarno je pozitivno (greška tipa II).  

Struktura matrice konfuzije na primeru binarnog klasifikatora prikazana je u narednoj tabeli.

|                    | Predviđeno pozitivno | Predviđeno negativno |
|--------------------|-------------------|-------------------|
| **Stvarno pozitivno**   | Istinski pozitivan (TP) | Lažno negativan (FN) |
| **Stvarno negativno**   | Lažno pozitivan (FP)   | Istinski negativni (TN) |

Matrica konfuzije omogućava detaljan uvid u tipove grešaka koje model pravi, kroz poređenje stvarnih i predviđenih klasa.

### 💳 Primer: Procena kreditnog rizika

|                       | **Predviđeno: Nizak rizik** | **Predviđeno: Visok rizik** |
|-----------------------|----------------------------|-----------------------------|
| **Stvarno: Nizak rizik** | Tačno pozitivni (TP)        | Lažno negativni (FN)         |
| **Stvarno: Visok rizik** | Lažno pozitivni (FP)        | Tačno negativni (TN)         |

---

## 🧮 Mere kvaliteta klasifikacije

Iz osnovnih vrednosti matrice konfuzije izvode se osnovne mere za evaluacije modela.

---

### ✅ Tačnost (engl. Accuracy)

**Tačnost** (engl. Accuracy) predstavlja osnovni indikator uspešnosti modela i definiše se kao definiše kao količnik broja ispravnih predviđanja i ukupnog broja instanci nad kojima je vršena klasifikacija.
#### 📐 Formula
U slučaju binarne klasifikacije, tačnost se izvodi iz elemenata matrice konfuzije primenom sledeće formule: 

$$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}.$$

#### 🧠 Tumačenje

Tačnost predstavlja procenat ukupnih uzoraka koje je model ispravno klasifikovao, bez obzira na to kojoj klasi pripadaju. 

Drugim rečima, tačnost odgovara na pitanje:  
> „Koliki je udeo ispravnih predviđanja modela u odnosu na ukupan broj analiziranih slučajeva?“


#### ⚠️ Ograničenja

U slučaju neuravnoteženih klasa (engl. imbalanced classes), tj. kada jedna klasa ima mnogo više primera od druge, visoka tačnost može biti zavaravajuća. Problem nastaje jer klasa sa više primera može dominirati modelom. Ako model uvek predviđa samo dominantnu klasu, tačnost može biti veštački visoka, jer se manje zastupljene klase ne uzimaju u obzir, što stvara iluziju dobre performanse modela. Zbog toga je neophodno koristiti i druge metrike koje koje prate uspešnost i za manje zastupljene klase.

---

### 🎯 Preciznost (engl. Precision)

**Preciznost** (engl. Precision) je mera koja pokazuje koliki deo instanci koje je model označio kao pozitivne zaista pripada pozitivnoj klasi. Drugim rečima, preciznost meri pouzdanost pozitivnih predikcija modela.

#### 📐 Formula

U slučaju binarne klasifikacije, preciznost se računa iz elemenata matrice konfuzije sledećom formulom:

$$\text{Precision} = \frac{TP}{TP + FP}.$$

#### 🧠 Tumačenje 

Preciznost se odnosi na pouzdanost pozitivnih predikcija modela i pokazuje koliki deo instanci koje je model označio kao pozitivne zaista pripada toj klasi. Drugim rečima, odgovara na pitanje: „Koliki procenat predviđenih pozitivnih instanci je tačan?“ Visoka preciznost znači da model retko greši kada proglasi neki primer pozitivnim.

Ova metrika dobija na značaju u scenarijima gde su posledice lažno pozitivnih grešaka kritične. Tipični primeri uključuju medicinsku dijagnostiku, gde pogrešna dijagnoza može dovesti do nepotrebnih medicinskih intervencija; sisteme za otkrivanje prevara, gde pogrešno blokiranje transakcija narušava profit; zatim i kod filtriranje neželjene pošte, gde greška nastaje kada važna poruka greškom bude klasifikovana kao neželjena.


#### ⚠️ Ograničenja

Slično ograničenjima tačnosti u kontekstu neuravnoteženih klasa, niska preciznost ukazuje na to da model pravi veliki broj lažno pozitivnih predikcija, odnosno označava negativne instance kao pozitivne. Visoka preciznost ne znači nužno da model dobro prepoznaje celinu pozitivne klase. Naime, model može biti veoma „oprezan“ i postizati visoku preciznost tako što će označiti samo nekoliko najsigurnijih primera, ali po cenu propuštanja mnogih stvarno pozitivnih uzoraka. Iz tog razloga, neophodno je preciznost interpretirati u sprezi sa drugim metrikama.  

---

## 🔄 Odziv (engl. Recall/Sensitivity)

**Odziv** (engl. Recall ili Sensitivity) meri sposobnost modela da pravilno prepozna sve instance pozitivne klase. Drugim rečima, odziv pokazuje koliki deo stvarno pozitivnih instanci je model uspešno detektovao.


### 📐 Formula
U slučaju binarne klasifikacije, odziv se računa iz elemenata matrice konfuzije sledećom formulom:

$$\text{Recall} = \frac{TP}{TP + FN}.$$

### 🧠 Tumačenje u finansijama

Model sa visokim odzivom uspešno prepoznaje najveći deo stvarno pozitivnih uzoraka. Ova metrika postaje prioritetna u situacijama kada propuštanje pozitivnih instanci nosi visoke rizike ili troškove. Tipični primeri uključuju medicinsku dijagnostiku, gde je od ključnog značaja detektovati bolest u ranoj fazi, kao i sisteme za otkrivanje prevara, gde je primarni cilj identifikovati što veći broj sumnjivih aktivnosti.

Odziv odgovara na pitanje:

> *Od svih uzoraka koji su zaista pozitivni, koliki procenat je model uspeo da pronađe i tačno identifikuje?*

### ⚠️ Ograničenja

Glavno ograničenje odziva je to što on ne uzima u obzir lažno pozitivne greške. To znači da model može imati visok odziv ako predviđa skoro sve pozitivne primere, ali pri tome pravi mnogo pogrešnih predikcija, što može dovesti do lažnog utiska o njegovoj efikasnosti. Zbog toga je preporučljivo koristiti odziv zajedno sa drugim metrikama.

---

## ⚖️ F1-mera (engl. F1-score)

**F1-mera** je definisan kao harmonijska sredina preciznosti i odziva, i koristi se kada je potrebno postići balans između ova dva kriterijuma, čime se dobija objedinjena metrika koja istovremeno vrednuje pouzdanost pozitivnih predviđanja i sveobuhvatnost prepoznavanja pozitivnih instanci. Ova mera je naročito značajna u radu sa neuravnoteženim klasama, jer sprečava preciznost ili odziv da samostalno pruže nerealnu sliku o performansama modela.

### 📐 Formula

Za binarnu klasifikaciju F1-mera se računa kao:

$$\text{F1-mera} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}.$$

### 🧠 Tumačenje

Visok F1-mera ukazuje na to da model simultano postiže visoke performanse u obe ključne dimenzije: on precizno identifikuje pozitivne instance, dok istovremeno uspešno detektuje većinu njih. Drugim rečima, model je podjednako efikasan u izbegavanju pogrešnih klasifikacija i u pronalaženju svih relevantnih primera.


## ⚖️ $F_\beta$-score

**$F_\beta$-score** je mera koja kombinuje preciznost i odziv u jednu jedinstvenu metriku, pri čemu parametar $\beta$ omogućava kontrolu relativne važnosti odziva u odnosu na preciznost. F1-score je posebni slučaj kada je $\beta$= 1, što znači da preciznost i odziv imaju jednak značaj.

#### 📐 Formula
$F_\beta$-score se računa sledećom formulom:

$$
F_\beta = (1 + \beta^2) \cdot \frac{\text{Precision} \cdot \text{Recall}}{(\beta^2 \cdot \text{Precision}) + \text{Recall}},
$$

gde je $\beta$ predstavlja parametar koji određuje koliko je odziv važniji ili manje važan u odnosu na preciznost.

#### 🧠 Tumačenje

- Kada je ** $\beta= 1$ **, F1-score daje jednak značaj preciznosti i odzivu, što predstavlja standardnu i najčešće korišćenu verziju. 
- Kada je **$\beta < 1$**, veći naglasak se stavlja na preciznost; model je strože kažnjen za lažno pozitivne predikcije nego za propuštanje stvarno pozitivnih instanci.
- Kada je **$\beta> 1$**, veći naglasak se stavlja na odziv; model je strože kažnjen za propuštanje stvarno pozitivnih instanci nego za lažno pozitivne predikcije.

Zahvaljujući ovoj fleksibilnosti, $F_\beta$-score je posebno pogodan za prilagođavanje modela specifičnim prioritetima problema.

#### ⚠️ Ograničenja

$F_\beta$-score, iako kombinuje dve važne metrike, ne odražava performanse modela po klasama pojedinačno kod ekstremno neuravnoteženih skupova podataka. U tim slučajevima može biti korisno kombinovati ga sa **matricom konfuzije** ili dodatnim metrikama poput **ROC-AUC**.

Iako $F_\beta$ mera kombinuje dve ključne metrike, on ne pruža uvid u performanse modela po pojedinačnim klasama kod ekstremno neuravnoteženih skupova podataka. U takvim situacijama korisno je koristiti ga zajedno sa drugim nabrojanim metrikama ili dodatnim metrikama poput **ROC-AUC**.



## 🏦 Praktični značaj u finansijskim primenama

U finansijskom sektoru, pravi izbor metrike evaluacije direktno utiče na **upravljanje rizicima** i **profitabilnost institucije**. Preciznost (Precision) se primenjuje u scenarijima gde bi **lažna uzbuna**, poput neosnovanog blokiranja kartice lojalnog klijenta, mogla narušiti reputaciju i korisničko iskustvo. Sa druge strane, odziv (Recall) postaje ključan kod detekcije kritičnih događaja, kao što su **pranje novca** ili **kreditni defolt**, gde propuštanje stvarno rizičnih slučajeva može imati znatno veće posledice od dodatne provere.  

Konačno, **F1-score** se nameće kao centralna mera u oblastima kao što su **algoritamsko trgovanje** i **odobravanje kredita**, jer omogućava balans između stroge selekcije profitabilnih prilika i sveobuhvatnog obuhvata tržišnih signala, sprečavajući model da postane previše konzervativan ili previše rizičan.
