# Nadgledano učenje

Nadgledano učenje predstavlja jedan od osnovnih pristupa u mašinskom učenju, u kome se model trenira na označenim podacima kako bi naučio da povezuje ulazne karakteristike sa odgovarajućim izlazima.

**Ishodi učenja**: Nakon proučavanja ovog poglavlja student će biti u stanju da:

- Objasni ključne pojmove i teorijske postavke nadgledanog učenja.
- Razlikuje osnovne tipove problema nadgledanog učenja, razlikujući zadatke klasifikacije od zadataka regresije.
- Interpretira matricu konfuzije i analizira performanse klasifikacionog modela.

## Uvod u nadgledano učenje

**Nadgledano učenje** (engl. *supervised learning*) je tip mašinskog učenja u okviru kojeg algoritam uči na skupu podataka u kome su poznati i ulazni podaci i odgovarajući izlazni podaci, poznati kao oznake (engl. labels).

Podaci za obuku sastoje se od parova **ulaz–izlaz** (npr. karakteristika i oznaka), zbog čega se ovakvi skupovi podataka često nazivaju **označenim skupovima podataka** (engl. labeled datasets).

Cilj algoritma nadgledanog učenja jeste da nauči **mapiranje (funkciju)** koja povezuje ulazne podatke sa odgovarajućim izlazima. Model se obučava na skupu podataka u kome su za zadate ulazne vrednosti poznate tačne izlazne vrednosti (jedna ili više), kako bi se uočilo pravilo preslikavanja i model osposobio da precizno predvidi rezultate za nove, ranije nekorišćene ulazne podatke.

Napomenimo da se polazi od pretpostavke da između ulaznih i izlaznih vrednosti postoji određena veza. Zadatak algoritma je da se trenira na postojećim podacima kako bi se ta veza identifikovala i iskoristila za predviđanje izlaznih vrednosti na osnovu novih ulaznih podataka.

---

**Ulazni podaci**

Svaka instanca u posmatranom skupu podataka predstavljena je kao *vektor ulaznih karakteristika*, odnosno atributa ili promenljivih.

Pojedinačna instanca $x_i$, $i\in\{1,2...,m\}$ iz skupa ulaznih podataka $X=\{x_1, x_2,..., x_m\}$ matematički se predstavlja u vektorskoj formi:


$$
x_i= (x_{i1}, x_{i2}, x_{i3}, \ldots, x_{in})
$$

gde svaka komponenta $x_{i1}, x_{i2}, x_{i3}, \ldots, x_{in}$ vektora predstavlja jednu karakteristiku instance $x_i$.

---

**Izlazni podaci**

Izlazne vrednosti se najčešće označavaju simbolom $y$ i predstavljaju **ciljnu promenljivu** (engl. *target variable*).

U većini slučajeva $y$ je jednodimenzionalna promenljiva, ali može biti i višedimenzionalna, u zavisnosti od prirode problema i tipa zadatka nadgledanog učenja.

Postoje i situacije u kojima ni $x$ ni $y$ nisu predstavljeni numeričkim vrednostima, već mogu imati složenije strukture, kao što su:
- sekvence,
- grafovi,
- tekstualni podaci,
- ili drugi oblici koji zahtevaju posebne pristupe obrade i modelovanja.

---

Nadgledano mašinsko učenje spada među najzastupljenije i u praksi najefikasnije pristupe u okviru mašinskog učenja. U narednim poglavljima biće detaljno razmotreni odabrani algoritmi iz ove grupe.

Tokom procesa treniranja modela mašinskog učenja, u okviru nadgledanog učenja, sistemu se obezbeđuju eksplicitni primeri za učenje, pri čemu su uz ulazne podatke dostupni i tačni izlazi. Ovaj proces je analogan situaciji u kojoj se učeniku prikazuju reprezentativni primeri uz jasno definisana objašnjenja, na osnovu kojih se postepeno formira razumevanje pravila i obrazaca.

**Primer iz analize kreditnog rizika**

Zamislimo da se algoritam obučava da preuzme ulogu specijaliste za procenu kreditnog rizika, oslanjajući se isključivo na istorijske podatke koji su unapred klasifikovani (obeleženi). Ovi podaci, koje nazivamo skupom za obuku, sastoje se od precizno definisanih parova:

- Ulazni vektor: Skup karakteristika koje opisuju profil klijenta i njegovu finansijsku istoriju (npr. mesečni prihodi, stabilnost zaposlenja, koeficijent zaduženosti, kreditna istorija, starost klijenta, nivo obrazovanja, svrha kredita, itd.),
- Izlaz koji predstavlja ishod, tj. status kredita (da li je kredit „uredan“ ili „problematičan“).

---

Model se oslanja na unapred dostupne (*apriori*) informacije kako binaučio funkciju koja povezuje ulaze sa odgovarajućim izlazima. Kvalitet podataka za obuku je od ključnog značaja, jer model nema sposobnost da sam prepozna i isključi netačne ili nepouzdane informacije, već u potpunosti gradi svoja pravila na osnovu onoga što mu je dato. Ukratko:

- ako su podaci visokog kvaliteta, model može donositi pouzdane odluke;
- ako su podaci nepouzdani ili sadrže greške, performanse i tačnost modela biće značajno ugrožene.


**Generalizacija i odnos složenosti modela**

Glavni cilj nadgledanog učenja je razvoj modela koji može da donosi tačne zaključke nad novim, prethodno neviđenim podacima, što se definiše kao sposobnost generalizacije. Da bismo razumeli kako se postiže ovaj cilj, neophodno je analizirati odnos između složenosti modela i njegove pouzdanosti.

* Suština generalizacije: Primarni zadatak modela je da prepozna osnovni obrazac (engl. pattern) u podacima, dok istovremeno mora ostati imun na šum i slučajne varijacije.
* Merodavnost evaluacije: Visoka tačnost postignuta na skupu za obuku nije konačno merilo uspeha. Prava vrednost i upotrebna moć modela dokazuju se isključivo njegovim performansama na test podacima.
* Problem nedovoljnog prilagođavanja: Ukoliko je model suviše jednostavan, on neće moći da obuhvati ključne zakonitosti unutar podataka, što rezultira lošim predviđanjima i na trening i na test skupu.
* Balans kompleksnosti i jednostavnosti: Iako složeniji modeli bolje opisuju podatke na kojima uče, oni su izrazito skloni prekomernom prilagođavanju (engl. overfitting).
* Princip jednostavnosti: U praksi se uvek teži izboru što jednostavnijeg modela koji je u stanju da obezbedi zadovoljavajuću preciznost uz očuvanje stabilne generalizacije.

> Što veću složenost dozvolimo modelu, to će on bolje predviđati na podacima za obuku. Međutim, ako model postane previše kompleksan, on počinje da „uči napamet” pojedinačne primere i šum iz trening skupa, čime nepovratno gubi sposobnost kvalitetne generalizacije na novim podacima.

## Osnovne kategorije modela nadgledanog učenja

Modele nadgledanog učenja možemo podeliti u dve glavne kategorije u zavisnosti od tipa izlaza koji želimo da dobijemo:

- **Klasifikacija** (Prepoznavanje obrazaca)  
  Model uči da svrstava opservacije u *diskretne kategorije* (klase). Njegov zadatak je da prepozna kojoj grupi posmatrana instanca pripada. Na primer, želimo da filtriramo elektronsku poštu i odredimo koji mejl je nepoželjna pošta, a koji regularna pošta. Model prima mejlove i uči da ih svrstava u ove dve grupe.   
  - Pitanje: „Da li je ovaj mejl nepoželjna pošta ili regularna pošta?“  
  - Izlaz: oznaka kategorije, na primer $0$ za regularnu poštu ili $1$ za nepoželjnu poštu.
  
- **Regresija** (Numeričko predviđanje)  
  Model uči da predviđa *kontinuirane numeričke vrednosti*, odnosno broj unutar određenog opsega, na osnovu uočenih obrazaca u podacima. Na primer, agencija za nekretnine želi da proceni cenu stanova. Na primer, želimo da predvidimo buduću cenu određene akcije na osnovu istorijskih podataka i tržišnih indikatora, kao što su indeksi volatilnosti, obim trgovanja, kamatne stope i prethodne cene akcija.
    - Pitanje: „Kolika će biti cena akcije u određenom vremenskom trenutku u budućnosti?“  
    - Izlaz: predviđena numerička vrednost cene akcije.


U okviru ovog udžbenika poseban akcenat stavljen je na algoritme klasifikacije. 
Razlog za to je činjenica da se veliki broj praktičnih problema u oblasti mašinskog učenja može 
formulisati kao zadatak razvrstavanja posmatranih instanci u unapred definisane kategorije. 
Tipični primeri takvih problema uključuju filtriranje elektronske pošte, prepoznavanje obrazaca 
na slikama, kao i različite zadatke iz oblasti medicinske dijagnostike i analize podataka.

Pored toga, klasifikacioni modeli omogućavaju primenu različitih metrika za procenu performansi 
modela, kao što su preciznost, odziv i F1-mera, čime se dobija detaljniji uvid u kvalitet 
izgrađenog modela. Iz navedenih razloga, u nastavku udžbenika veća pažnja posvećena je analizi 
i primeni algoritama klasifikacije u okviru nadgledanog učenja.

> **Napomena o univerzalnosti algoritama** > Neophodno je istaći da značajan broj algoritama poseduje dualnu prirodu, nudeći specifične implementacije kako za probleme klasifikacije, tako i za probleme regresije. Savladavanje fundamentalnih principa rada na primerima klasifikacije obezbeđuje teorijsku i praktičnu osnovu koja je direktno transferabilna na regresione modele. Ključna razlika ogleda se primarno u prirodi izlaznih veličina, prelasku sa diskretnih kategorijalnih klasa na kontinualne numeričke vrednosti.

Najjednostavniji način razlikovanja klasifikacije i regresije zasniva se na prirodi izlazne promenljive:

* Ukoliko izlazna promenljiva može da poprimi kontinuirane vrednosti, odnosno postoji određen interval mogućih ishoda, tada se radi o regresionom problemu. 
* Ukoliko su izlazne vrednosti diskretne kategorije, reč je o klasifikaciji. U ovom slučaju, cilj modela je da objekte rasporedi u unapred jasno definisane grupe.

**Algoritmi klasifikacije** predstavljaju podvrstu nadgledanog učenja, i jednu od najznačajnijih i najšire zastupljenih oblasti mašinskog učenja. Cilj klasifikacije je da se, učeći na primerima sa poznatim ishodima iz označenog skupa podataka za treniranje, pronađe funkcija koja preslikava ulazne podatke u odgovarajuće kategorije. Takva funkcija omogućava predikciju oznake klase za nove, neviđene instance, tako što ih, na osnovu njihovih atributa (karakteristika), svrstava u jednu od unapred definisanih grupa koje nazivamo oznake. U procesu klasifikacije, izlazna promenljiva $y$ označava pripadnost određenoj klasi ili kategoriji. U procesu klasifikacije, izlazna promenljiva $y$ označava pripadnost određenoj *klasi ili kategoriji*, formalno:  

$$y \in \{1, 2, \dots, C\},$$

gde je $C$ broj unapred definisanih klasa. Svaka izlazna vrednost, odnosno oznaka, pripada jednoj od unapred definisanih klasa koje predstavljaju diskretne i međusobno isključive kategorije. Shodno tome, svaka nova instanca koju model obrađuje mora biti klasifikovana u tačno jednu od $C$ raspoloživih klasa. 

Ulazni podaci, odnosno atributi, mogu biti različitih tipova. Najčešće se razlikuju numerički (npr. realne vrednosti, intenzitet piksela, fizička merenja, itd.) i kategorički atributi (npr. pol, tip zemljišta, nivo obrazovanja, itd.).

**Tipovi klasifikacije u nadgledanom učenju**

Klasifikacija je proces predviđanja diskretne oznake klase iz unapred definisanog skupa mogućih vrednosti. U zavisnosti od broja kategorija, klasifikacione zadatke delimo na tri osnovne grupe:

- **Binarna klasifikacija**  
Ovo je poseban slučaj gde model bira između *tačno dve klase*. Najlakše ju je razumeti kao davanje odgovora na *da/ne* pitanja. Najčešće se formalno zapisuje kao $y \in \{0, 1\}$ ili $y \in \{-1, 1\}$. Kao tipičan primer binarne klasifikacije možemo uzeti detekciju neželjene pošte. U ovom slučaju model treba da utvrdi da li određena poruka pripada kategoriji neželjene ili legitimne pošte, pri čemu postoje samo dva moguća ishoda. Ako se utvrdi da je poruka neželjena, ona se svrstava u pozitivnu klasu, dok se u suprotnom klasifikuje kao legitimna pošta, odnosno negativna klasa.
    Dodatni primeri:
    * Medicinska dijagnostika: da li je nalaz „pozitivan“ ili „negativan“?
    * Odobravanje kredita: da li se zahtev „prihvata“ ili „odbija“?
    

- **Višeklasna klasifikacija**   
Primenjuje se kada je potrebno izvršiti razvrstavanje u *više od dve* međusobno isključive klase. U ovom slučaju ciljna promenljiva poprima vrednost iz skupa $y \in \{1, 2, \dots, C\}$, gde je $C > 2$ (na primer, kod prepoznavanja rukom pisanih cifara od 0 do 9). Kod standardne višeklasne klasifikacije klase su međusobno nezavisne i ne postoji prirodan redosled među njima. Drugim rečima, model uči kojoj kategoriji posmatrana instanca pripada, bez pretpostavke da je neka klasa „veća“, „manja“ ili „bolja“ od druge. 
    Primeri:
    * Računarski vid: prepoznavanje životinja na fotografiji (npr. *mačka*, *pas*, *ptica*, *medved*);
    * Kategorizacija teksta: automatsko svrstavanje vesti u rubrike (npr. *politika*, *sport*, *ekonomija*, *zabava*);
    * Prepoznavanje rukopisa: klasifikacija cifara od 0 do 9. 

Međutim, u određenim situacijama kategorije mogu imati jasno definisan prirodan ili preferencijalni redosled. Takav pristup naziva se *ordinalna klasifikacija* (engl. ordinal classification). Kod ovih problema klase imaju svoj redosled, ali razmak između njih nije precizno kvantifikovan kao kod numeričkih vrednosti. Primeri ordinalnih kategorija uključuju nivo zadovoljstva korisnika (nizak, srednji, visok), školske ocene (1, 2, 3, 4, 5), kao i nivoe stručnosti (početni, srednji, napredni).


- **Klasifikacija sa više oznaka** 
Podrazumeva situaciju u kojoj jedan podatak istovremeno može biti dodeljen većem broju klasa, jer njegove karakteristike odgovaraju različitim kategorijama.
    Primeri:
    * Kategorizacija vesti: Jedan članak može biti označen kao politika, ekonomija i međunarodni odnosi istovremeno.
    * Analiza slika: Fotografija može sadržati više oznaka, npr. osoba, drvo i more.
    * Preporuka muzike: Jedna pesma može pripadati žanrovima Pop, Elektronska i Dance.
    
---


| Tip klasifikacije | Broj klasa | Primer pitanja |
| :--- | :--- | :--- |
| **Binarna** | Tačno 2 | Da li je transakcija regularna ili sumnjiva? |
| **Višeklasna** | Više od 2 (bira se jedna) | U koju kategoriju kreditnog rizika spada klijent? |
| **Sa više oznaka** | Više od 2 (bira se više istovremeno) | Koje se sve teme pominju u ovom članku (npr. politika, sport i ekonomija)? |

> **Napomena:** Kod binarne klasifikacije često jednu klasu nazivamo "pozitivnom" (ona koju želimo da detektujemo), a drugu "negativnom", bez obzira na njihovu semantičku vrednost. Pozitivna klasa je obično ona koja predstavlja prisustvo određenog stanja, događaja ili anomalije koju model treba da izoluje. Na primer, u zadacima detekcije neželjene pošte, pozitivna klasa se često odnosi na spam poruke. Izbor pozitivne klase nije univerzalan, već predstavlja subjektivnu odluku koja zavisi od oblasti primene.

**Primena klasifikacionih algoritama** obuhvata mnoge savremene servise i sisteme, često neprimetno za korisnike, gde obavljaju zadatke kao što su:

* Automatskog filtriranja nepoželjne elektronske pošte,  
* Detekcije prevara u finansijskim transakcijama,  
* Procene kreditnog rizika,  
* Biometrijske identifikacije pri otključavanju pametnih uređaja, 
* Automatske kategorizacije i rangiranja sadržaja na digitalnim platformama,
* Sistemi za preporuku: klasifikacija korisnika ili sadržaja (npr. procena da li će se korisniku dopasti određeni proizvod).


**Ilustracija klasifikacionih modela kroz primere iz bankarskog sektora**

U okviru sistema za upravljanje rizicima, svakodnevno se analizira veliki broj zahteva za odobravanje kredita. Za svaki zahtev neophodna je evaluacija kreditne sposobnosti klijenta, odnosno verovatnoće da će preuzete obaveze biti uredno izmirene.

Razmotrićemo tri scenarija:

* Scenario 1 (binarna klasifikacija):  Klijent podnosi zahtev za stambeni kredit, a model analizira njegove ključne atribute, kao što su mesečna primanja, kreditna istorija i radni staž. Na osnovu obrazaca identifikovanih u skupu za treniranje, model sprovodi binarnu klasifikaciju gde se klijent mapira u jednu od dve diskretne kategorije:
     * Nizak rizik podrazumeva odobravanje zahteva klijenta
     * Visok rizik podrazumeva odbijanje zahteva klijenta


* Scenario 2 (višeklasna klasifikacija): U složenim finansijskim sistemima, odlučivanje se retko zasniva na binarnim ishodima; umesto toga, klijenti se procenjuju kroz sistem kreditnog rejtinga. Analizom finansijskih pokazatelja, kao što su profitabilnost, likvidnost i stabilnost prihoda, model razvrstava klijente u unapred definisane klase:
    * Klasa A predstavlja investicioni rang koji odlikuje visok nivo sigurnosti i stabilnosti,
    * Klasa B obuhvata špekulativni rang sa umerenim rizikom i većom osetljivošću na tržišne promene,
    * Klasa C označava visok rizik uz povećanu verovatnoću neizmirenja obaveza ili bankrota.

* Scenario 3 (klasifikacija sa više oznaka): Banke često koriste modele kako bi istovremeno identifikovale više različitih karakteristika klijenta na osnovu istorije transakcija. U ovom slučaju, jedan klijent može istovremeno dobiti više oznaka (etiketa) koje nisu međusobno isključive:
    * Korisnik digitalnog bankarstva (primarno koristi mobilnu aplikaciju),
    * Potencijalni investitor (često trguje hartijama od vrednosti),
    * Putnik (često troši sredstva u inostranstvu).

Preciznost ovakvih klasifikacionih modela ima neposredan uticaj na kapitalni bilans i strateško planiranje finansijskih institucija. Na primer, pogrešna alokacija visokorizičnog klijenta u kategoriju investicionog ranga (Klasa A) može uzrokovati ozbiljne finansijske gubitke usled neizmirenja obaveza.

U nastavku ovog poglavlja ćemo predstaviti ključne metrike za evaluaciju performansi i pouzdanosti ovakvih klasifikatora.


### Algoritmi za klasifikaciju 

Za rešavanje problema klasifikacije postoji širok spektar algoritama mašinskog učenja (machine learning algorithms).  
Među najzastupljenijima su: 

- Logistička regresija (engl. Logistic Regression)  
- K-najbližih suseda (engl. K-Nearest Neighbors, k-NN)  
- Stabla odlučivanja (engl. Decision Trees)  
- Mašine sa potpornim vektorima (engl. Support Vector Machines, SVM)  
- Naivni Bajes (engl. Naive Bayes)  
- Ansambl metode (engl. Ensemble Methods)  
- Slučajne šume (engl. Random Forest)  
- Gradient Boosting (engl. Gradient Boosting Machines, GBM)  
- Veštačke neuronske mreže (engl. Artificial Neural Networks, ANN)  
- XGBoost (engl. Extreme Gradient Boosting)  
- LightGBM (engl. Light Gradient Boosting Machine)  
- Kombinacije ansambla (engl. Stacking / Voting Classifier)

**Uporedni pregled algoritama nadgledanog učenja**


U nastavku je prikazan sistematizovan pregled najznačajnijih algoritama koji se primenjuju u savremenom mašinskom učenju. Imajući u vidu da ne postoji univerzalno najbolji algoritam (u skladu sa principom *'No Free Lunch'*), izbor odgovarajućeg modela zasniva se na razumevanju njegovih karakteristika i specifičnosti problema koji se rešava.

Iz ove uporedne analize možemo uočiti nekoliko ključnih aspekata:

   * Složenost naspram interpretabilnosti: Algoritmi poput logističke regresije i stabala odlučivanja nude jasnu logiku odlučivanja, dok napredne metode poput neuronskih mreža i ansambala pružaju veću preciznost po cenu teže interpretacije rezultata.
   * Računarska efikasnost: Naivni Bajes i algoritam k-najbližih suseda se izdvajaju po brzini, što ih čini pogodnim za početne faze istraživanja, dok algoritmi zasnovani na gradijentnom pojačanju zahtevaju značajnije hardverske resurse i pažljivo podešavanje hiperparametara.
   * Robusnost na podatke: Vidljivo je da su određeni modeli, poput mašine potpornih vektora i slučajnih šuma, dizajnirani da minimizuju rizik od prekomernog prilagođavanja, dok drugi zahtevaju rigoroznu pretobradu podataka (npr. skaliranje kod algoritma k-najbližih suseda).

Ovaj pregled služi kao orijentir prilikom izbora adekvatne metode za konkretan problem, balansirajući između zahteva za preciznošću, brzinom i mogućnošću objašnjenja odluka modela.

| Model | Opis | Prednosti | Mane |
| :--- | :--- | :--- | :--- |
| **Logistička regresija** | Statistički model koji predviđa verovatnoću pripadnosti klasi koristeći sigmoidnu funkciju. | Jednostavan, brz, lako interpretabilan. | Pretpostavlja linearnost; loše modeluje složene nelinearne odnose. |
| **k-najbližih suseda** | Klasifikuje instance na osnovu većinske klase njenih $k$ najbližih suseda u prostoru podataka. | Intuitivan, ne zahteva fazu treniranja. | Spor kod velikih skupova podataka; osetljiv na autlajere i skalu podataka. |
| **Stabla odlučivanja** | Hijerarhijska struktura koja donosi odluke kroz niz logičkih testova nad atributima. | Laka vizuelizacija i interpretacija; radi sa različitim tipovima podataka. | Sklona prekomernom prilagođavanju (*overfitting*); nestabilna na male promene. |
| **Mašine potpornih vektora** | Pronalazi optimalnu hiperravan koja maksimizuje marginu između klasa u višedimenzionalnom prostoru. | Efikasan u prostorima visokih dimenzija; otporan na prekomerno prilagođavanje. | Zahteva pažljivo podešavanje hiperparametara; računarski zahtevan za velike baze. |
| **Naivni Bajes** | Verovatnoćnosni model zasnovan na Bajesovoj teoremi uz pretpostavku nezavisnosti atributa. | Izuzetno brz; efikasan kod klasifikacije teksta i velikih skupova. | Pretpostavka nezavisnosti atributa često nije ispunjena u realnosti. |
| **Slučajne šume** | Ansambl metod koji kombinuje veći broj stabala odlučivanja radi povećanja tačnosti. | Visoka robusnost i tačnost; smanjuje rizik od preprilagođavanja. | Teže se interpretira u poređenju sa pojedinačnim stablom; sporiji rad. |
| **XGBoost / LightGBM** | Napredni algoritmi zasnovani na gradijentnom pojačanju optimizovani za brzinu i performanse. | Vrhunska preciznost; efikasno rukovanje nedostajućim vrednostima. | Kompleksni za podešavanje; mogu biti skloni prekomernom prilagođavanju ako se ne regulišu. |
| **Neuronske mreže** | Modeli inspirisani biološkim mozgom, sastavljeni od slojeva neurona za učenje složenih obrazaca. | Moćni za nelinearne probleme (slike, zvuk, tekst). | "Crna kutija" (teška interpretacija); zahtevaju ogromne resurse i podatke. |
| **Stacking / Voting** | Kombinovanje različitih tipova modela kako bi se iskoristile njihove pojedinačne snage. | Maksimizuje prediktivnu moć; smanjuje varijansu i pristrasnost. | Visoka kompleksnost implementacije; zahtevno vreme izvršavanja. |

Izbor algoritma zavisi od specifičnosti problema, kao i od strukture, obima i kompleksnosti samih podataka, budući da svaki model ima svoje prednosti i mane.

### Metrike za evaluaciju uspešnosti klasifikacionih modela

Evaluacija modela vrši se na osnovu njegove sposobnosti da **tačno predvidi klasu novih, do tada neviđenih podataka**.  
Kod problema klasifikacije, kvalitet modela se ne može uvek pouzdano proceniti samo ukupnom tačnošću, naročito u slučajevima kada su klase **neuravnotežene**  (engl. imbalanced classes), što je česta pojava u finansijskim podacima. Tipičan primer neuravnoteženih klasa u  podacima iz prakse je detekcija prevara pri korišćenju platnih kartica. U realnim finansijskim setovima podataka, ogromna većina transakcija je legitimna, dok prevare čine tek zanemarljiv procenat ukupnog uzorka. Ovakva distribucija dovodi do pojave klasne neravnoteže, gde jedna kategorija (većinska klasa) numerički dominira nad drugom (manjinskom klasom). Glavni izazov u ovakvim scenarijima je što model može ostvariti prividno visoku tačnost jednostavnim predviđanjem većinske klase za svaki ulazni podatak. Na primer, ako je 99% transakcija legitimno, model koji svaku transakciju označi kao ispravnu biće 99% tačan, ali potpuno beskoristan jer neće detektovati nijednu prevaru. Upravo zato se kod neuravnoteženih podataka primenjuju specifične tehnike evaluacije i napredne metode uzorkovanja.

Zbog toga se koriste mere zasnovane na matrici konfuzije.

---

**Matrica konfuzije**


Matrica konfuzije (engl. Confusion Matrix) se koristi za procenu performansi klasifikacionog modela mašinskog učenja.  
Ona tabelarno sistematizuje broj tačnih i pogrešnih predviđanja. Na osnovu ovih podataka izračunavaju se ključne metrike za ocenu kvaliteta samog modela, što omogućava analizu uspešnosti klasifikacije.  

Kod binarne klasifikacije, matrica konfuzije prikazuje četiri osnovna ishoda:

- Istinski pozitivan (engl. True Positive, TP): model je predvideo pozitivno, a stvarno je pozitivno.  
- Istinski negativni (engl. True Negative, TN): model je predvideo negativno, a stvarno je negativno.  
- Lažno pozitivan (engl. False Positive, FP): model je predvideo pozitivno, a stvarno je negativno (greška tipa I).  
- Lažno negativan (engl. False Negative, FN): model je predvideo negativno, a stvarno je pozitivno (greška tipa II).  

Struktura matrice konfuzije na primeru binarnog klasifikatora prikazana je u narednoj tabeli.



|                    | Predviđeno pozitivno | Predviđeno negativno |
|--------------------|-------------------|-------------------|
| **Stvarno pozitivno**   | Istinski pozitivan (TP) | Lažno negativan (FN) |
| **Stvarno negativno**   | Lažno pozitivan (FP)   | Istinski negativni (TN) |



Matrica konfuzije omogućava detaljan uvid u tipove grešaka koje model pravi, kroz poređenje stvarnih i predviđenih klasa.


**📊 Mere kvaliteta klasifikacije**

Iz osnovnih vrednosti matrice konfuzije izvode se osnovne mere za evaluaciju modela.

---

**Tačnost** 

*Tačnost* (engl. Accuracy) predstavlja osnovni indikator uspešnosti modela i definiše se kao količnik broja ispravnih predviđanja i ukupnog broja instanci nad kojima je vršena klasifikacija.

U slučaju binarne klasifikacije, tačnost se izvodi iz elemenata matrice konfuzije primenom sledeće formule: 

$$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}.$$

Tačnost predstavlja procenat ukupnih uzoraka koje je model ispravno klasifikovao, bez obzira na to kojoj klasi pripadaju. 

Drugim rečima, tačnost odgovara na pitanje:  
> „Koliki je udeo ispravnih predviđanja modela u odnosu na ukupan broj analiziranih slučajeva?“


U slučaju neuravnoteženih klasa (engl. imbalanced classes), tj. kada jedna klasa ima mnogo više primera od druge, visoka tačnost može biti zavaravajuća. Problem nastaje jer klasa sa više primera može dominirati modelom. Ako model uvek predviđa samo dominantnu klasu, tačnost može biti veštački visoka, jer se manje zastupljene klase ne uzimaju u obzir, što stvara iluziju dobre performanse modela. Zbog toga je neophodno koristiti i druge metrike.

---

**Preciznost**

*Preciznost* (engl. Precision) je mera koja pokazuje koliki deo instanci koje je model označio kao pozitivne zaista pripada pozitivnoj klasi. Drugim rečima, preciznost meri pouzdanost pozitivnih predikcija modela.


U slučaju binarne klasifikacije, preciznost se računa iz elemenata matrice konfuzije sledećom formulom:

$$\text{Precision} = \frac{TP}{TP + FP}.$$


Preciznost se odnosi na pouzdanost pozitivnih predikcija modela i pokazuje koliki deo instanci koje je model označio kao pozitivne zaista pripada toj klasi. Drugim rečima, odgovara na pitanje: „Koliki procenat predviđenih pozitivnih instanci je tačan?“ Visoka preciznost znači da model retko greši kada proglasi neki primer pozitivnim.

Ova metrika dobija na značaju u scenarijima gde su posledice lažno pozitivnih grešaka kritične. Tipični primeri uključuju medicinsku dijagnostiku, gde pogrešna dijagnoza može dovesti do nepotrebnih medicinskih intervencija; sisteme za otkrivanje prevara, gde pogrešno blokiranje transakcija narušava profit; zatim i kod filtriranje neželjene pošte, gde greška nastaje kada važna poruka greškom bude klasifikovana kao neželjena.


Slično ograničenjima tačnosti u kontekstu neuravnoteženih klasa, niska preciznost ukazuje na to da model pravi veliki broj lažno pozitivnih predikcija, odnosno označava negativne instance kao pozitivne. Visoka preciznost ne znači nužno da model dobro prepoznaje celinu pozitivne klase. Naime, model može biti veoma „oprezan“ i postizati visoku preciznost tako što će označiti samo nekoliko najsigurnijih primera, ali po cenu propuštanja mnogih stvarno pozitivnih uzoraka. Iz tog razloga, neophodno je preciznost interpretirati u sprezi sa drugim metrikama.  

---

**Odziv** 

*Odziv* (engl. Recall ili Sensitivity) meri sposobnost modela da pravilno prepozna sve instance pozitivne klase. Drugim rečima, odziv pokazuje koliki deo stvarno pozitivnih instanci je model uspešno detektovao.


U slučaju binarne klasifikacije, odziv se računa iz elemenata matrice konfuzije sledećom formulom:

$$\text{Recall} = \frac{TP}{TP + FN}.$$


Model sa visokim odzivom uspešno prepoznaje najveći deo stvarno pozitivnih uzoraka. Ova metrika postaje prioritetna u situacijama kada propuštanje pozitivnih instanci nosi visoke rizike ili troškove. Tipični primeri uključuju medicinsku dijagnostiku, gde je od ključnog značaja detektovati bolest u ranoj fazi, kao i sisteme za otkrivanje prevara, gde je primarni cilj identifikovati što veći broj sumnjivih aktivnosti.

Odziv odgovara na pitanje:

> *Od svih uzoraka koji su zaista pozitivni, koliki procenat je model uspeo da pronađe i tačno identifikuje?*

Glavno ograničenje odziva je to što on ne uzima u obzir lažno pozitivne greške. To znači da model može imati visok odziv ako predviđa skoro sve pozitivne primere, ali pri tome pravi mnogo pogrešnih predikcija, što može dovesti do lažnog utiska o njegovoj efikasnosti. Zbog toga je preporučljivo koristiti odziv zajedno sa drugim metrikama.

---

**F1-mera** 

*F1-mera* (engl. F1-score) je definisan kao harmonijska sredina preciznosti i odziva, i koristi se kada je potrebno postići balans između ova dva kriterijuma, čime se dobija objedinjena metrika koja istovremeno vrednuje pouzdanost pozitivnih predviđanja i sveobuhvatnost prepoznavanja pozitivnih instanci. Ova mera je naročito značajna u radu sa neuravnoteženim klasama, jer sprečava preciznost ili odziv da samostalno pruže nerealnu sliku o performansama modela.


Za binarnu klasifikaciju F1-mera se računa kao:

$$\text{F1-mera} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}.$$


Visok F1-mera ukazuje na to da model simultano postiže visoke performanse u obe ključne dimenzije: on precizno identifikuje pozitivne instance, dok istovremeno uspešno detektuje većinu njih. Drugim rečima, model je podjednako efikasan u izbegavanju pogrešnih klasifikacija i u pronalaženju svih relevantnih primera.


**$F_\beta$-mera**

*$F_\beta$-mera* (engl. $F_\beta$-score) je mera koja kombinuje preciznost i odziv u jednu jedinstvenu metriku, pri čemu parametar $\beta$ omogućava kontrolu relativne važnosti odziva u odnosu na preciznost. F1-mera je posebni slučaj kada je $\beta$= 1, što znači da preciznost i odziv imaju jednak značaj.

$F_\beta$-mera se računa sledećom formulom:

$$
F_\beta = (1 + \beta^2) \cdot \frac{\text{Precision} \cdot \text{Recall}}{(\beta^2 \cdot \text{Precision}) + \text{Recall}},
$$

gde je $\beta$ predstavlja parametar koji određuje koliko je odziv važniji ili manje važan u odnosu na preciznost.

- Kada je **$\beta= 1$**, F1-score daje jednak značaj preciznosti i odzivu, što predstavlja standardnu i najčešće korišćenu verziju. 
- Kada je **$\beta < 1$**, veći naglasak se stavlja na preciznost; model je strože kažnjen za lažno pozitivne predikcije nego za propuštanje stvarno pozitivnih instanci.
- Kada je **$\beta> 1$**, veći naglasak se stavlja na odziv; model je strože kažnjen za propuštanje stvarno pozitivnih instanci nego za lažno pozitivne predikcije.

Zahvaljujući ovoj fleksibilnosti, $F_\beta$-score je posebno pogodan za prilagođavanje modela specifičnim prioritetima problema.

Iako $F_\beta$ mera kombinuje dve ključne metrike, on ne pruža uvid u performanse modela po pojedinačnim klasama kod ekstremno neuravnoteženih skupova podataka. U takvim situacijama korisno je koristiti ga zajedno sa drugim metrikama.


---


**Metrike za evaluaciju višeklasnih klasifikatora**

Matrica konfuzije za višeklasne probleme omogućava detaljan uvid u performanse modela tako što mapira predviđene vrednosti (kolone) naspram stvarnih klasa (redovi). Na glavnoj dijagonali matrice nalaze se ispravno klasifikovani primeri (tamo gde se predviđanje poklapa sa stvarnošću), dok vrednosti van dijagonale direktno ukazuju na broj i tip grešaka.

Kod problema višeklasne klasifikacije često je potrebno objediniti rezultate dobijene za sve klase kako bi se dobila jedinstvena procena ukupnih performansi modela. U tu svrhu koriste se različiti načini *agregiranja metrika*, kao što su makro prosek, mikro prosek i težinski prosek. Pristup biramo prema specifičnim potrebama našeg skupa podataka i cilju koji želimo da postignemo.

**Makro prosek**  

Makro prosek (engl. macro average) se fokusira na ravnopravnost klasa, pa se vrednost metrike (npr. precizbost, tačnots, odziv, recall ili F1-mera) najpre izračunava za svaku klasu pojedinačno, nakon čega se računa njihov aritmetički prosek, koji predstavlja *makro* vrednost te metrike. Dakle, bez obzira na to koliko primera u bazi pripada pojedinačnoj klasi, svaka klasa ima jednak uticaj na konačnu vrednost metrike.


**Mikro prosek**  

Mikro prosek (engl. micro average) daje veći značaj klasama koje imaju veći broj instanci u skupu podataka. Vrednost metrike se računa globalno, tako što se najpre saberu ukupne vrednosti tačnih pozitivnih, netačnih pozitivnih i netačnih negativnih predikcija za sve klase, a zatim se na osnovu tih zbirnih vrednosti izračunava odgovarajuća metrika.


**Težinski prosek**  

Težinski prosek (engl. weighted average) balansira između makro i mikro pristupa. Metrika se prvo računa za svaku klasu pojedinačno, a zatim se te vrednosti ponderišu u skladu sa zastupljenošću svake klase u skupu podataka. Na ovaj način, klase sa većim brojem instanci imaju proporcionalno veći uticaj na krajnji rezultat, što težinski prosek čini veoma pouzdanim pokazateljem kod *neuravnoteženih skupova podataka*.


Glavne karakteristike i primena metoda za agregiranje metrika sumirani su u tabeli ispod.


| Način agregiranja | Metodologija | Uticaj raspodele klasa | Kada se koristi |
|-------------------|-----------------|------------------------|-----------------|
| **Makro prosek** | Metrika se računa za svaku klasu posebno, a zatim se uzima aritmetički prosek svih klasa. | Ravnopravan značaj: Svaka klasa vredi isto. | Kada je cilj podjednako dobar učinak na svim klasama. |
| **Mikro prosek** | Sabiraju se ukupni TP, FP i FN za sve klase, a zatim se izračunava metrika. | Veći uticaj imaju klase sa više primera u bazi. | Kada je bitna opšta preciznost modela na celokupnom uzorku podataka. |
| **Težinski prosek** | Metrika se računa za svaku klasu, a zatim se uzima prosek ponderisan brojem instanci svake klase. | Klase sa **više primera imaju veći uticaj**. | Kod neuravnoteženih podataka gde želimo balans između važnosti klase i njene veličine. |

**Izračunavanje metrika po klasama**

Za svaku klasu računamo preciznost i odziv koristeći One-vs-Rest pristup.

*Klasa: Nizak rizik*
* TP: 70
* FP: 10 + 1 = 11
* FN: 8 + 2 = 10
* Preciznost = 70 / 81 ≈ 0.86
* Odziv = 70 / 80 = 0.875 (88%)

*Klasa: Srednji rizik*
* TP: 22
* FP: 8 + 4 = 12
* FN: 10 + 8 = 18
* Preciznost = 22 / 34 ≈ 0.65
* Odziv = 22 / 40 = 0.55

*Klasa: Visok rizik*
* TP: 5
* FP: 2 + 8 = 10
* FN: 1 + 4 = 5
* Preciznost = 5 / 15 ≈ 0.33
* Odziv = 5 / 10 = 0.50

| Mikro prosek | (70 + 22 + 5) / 130 | 0.75 |

Težinski prosek pruža optimističniju procenu, jer je on pod direktnim uticajem najučestalije klase u podacima, gde model ostvaruje najbolje rezultate.

Pravi izbor metrike evaluacije direktno utiče na upravljanje rizicima i profitabilnost institucije. Preciznost se primenjuje u scenarijima gde bi lažna uzbuna, poput neosnovanog blokiranja kartice lojalnog klijenta, mogla narušiti reputaciju i korisničko iskustvo.
