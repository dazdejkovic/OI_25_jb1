# Upravljanje greškama

Greške u kodiranju u Pajtonu nisu znak da „ne znaš ništa“, nego da zaista radiš sa kodom. Svaki programer – od početnika do profesionalca – provodi mnogo vremena baš u ispravljanju grešaka.

Kada napišeš Python kod i pritisneš „run“, mogu da se dese tri glavne vrste grešaka:

- **Sintaksne greške**. To su greške u „gramatici“ jezika. Na primer: zaboravljena dvotačka, pogrešan broj zagrada, loš razmak kod uvlačenja (indentation). Pajton tada odmah prijavljuje grešku i uopšte ne pokreće program.

- **Greške u izvršavanju (runtime greške)** Kod je „pravilno napisan“, ali se pri izvršavanju desi nešto neočekivano. Na primer: deljenje sa nulom, pristup elementu liste koji ne postoji, korišćenje promenljive koja nije definisana. Program krene, ali se sruši u toku rada i Python izbaci Traceback“ sa opisom problema.

- **Logičke greške**. Najpodmuklije 🙂. Kod radi, nema poruka o grešci – ali rezultati nisu ono što očekuješ. Na primer: pogrešna formula, pogrešno poređenje, petlja koja broji „od 0 do 9“, a ti si mislio da broji „od 1 do 10“. Njih Pajton ne vidi – moraš ti da testiraš i razmišljaš.

U praksi se početnici u Pajtonu najčešće muče sa: 
- zaboravljenom dvotačkom (if, for, while, def, class)
- pogrešnim uvlačenjem (jedan red uvucen razmakom, drugi tabom)
- mešanjem tipova ("5" + 3)
- pogrešnim imenom promenljive (rezultat vs rezultati)
- off-by-one greškama u petljama (krećeš od 0, a misliš da krećeš od 1)

Zato je čitanje poruke o grešci jedna od najvažnijih veština. Pajton gotovo uvek da sledeće informacije:
- koja linija je problematična
- o kojoj vrsti greške se radi (SyntaxError, TypeError, NameError, IndexError, …)
- kratko objašnjenje šta mu smeta

U uvodu o greškama u kodiranju cilj je da student shvati dve stvari:
- Greške su normalan i očekivan deo programiranja, ne neuspeh.
- Umesto da paniči, treba polako da: pročita poruku o grešci, pronađe navedenu liniju, pokuša da razume zašto se to desilo, i korak po korak ispravi problem.

## Logičke greške


Jedna od grešaka koje se najteže pronalaze. Logička greška neće izbaciti poruku o grešci. Iako će program raditi glatko, postoji greška jer na izlazu na dobijamo rešenje koje očekujemo. Na primer, posmatrajmo sledeću nekorektnu primenu funkcije faktorijel.

````python
def los_faktorijel(n):
    izlaz = 1
    for k in range(1,n):
        izlaz*=k
    return izlaz
````

````python
los_faktorijel(5)
````

## Naredba try/except

Često je važno napisti program koji će upravljati odgovarajućim tipovima grešaka ili izuzetaka. Ili još konkretnije, greška ili izuzetak ne mora da uzrokuje kritičnu grešku takvu da vaš program prestane sa izvršavanjem. Naredba **try-except** je blok koda koji omogućava da naš program preduzme alternativne akcije u slučaju nastanka neke greške.   


**KONSTRUKCIJA**
```python
try:
    blok koda 1
except Ime_izuzetka:
    blok koda 2
```

:::{admonition} Primer 2.
:class: tip

Sledeći kod pokriva grešku kada programer napravi grešku sa tipovima podataka.

```python
a = [2]
try:
    if a <= 4:
        print(f"Broj {a} je manji ili jednak od 4")
except TypeError:
    print(f"Greška! {a} nije broj. Pokušajte ponovo.")
```
Greška! [2] nije broj. Pokušajte ponovo.
:::

Međutim ako korisnik prilikom izvršavanja sledećeg koda umesto broja unese na primer neka slova, try-except neće upravljati greškom:

````python
a = int(input("Unesite broj: "))
try:
    if a <= 4:
        print(f"Broj {a} je manji ili jednak od 4")
except TypeError: # Prebacicemo ovo u ValueError
    print(f"Greška! {a} nije celi broj. Pokušajte ponovo.")
````

Zašto try-except ne funkcioniše?

```{admonition} Važno:
:class: important 
Ako vaše upravljanje greškom pokuša da pokrije izuzetkom tipa koji ne hvata `except`, tada nastaje greška i izvršavanje se prekida.
```

````python
a  = [2]
try:
    if a <= 4:
        print(f"Broj {a} je manji ili jednak od 4")
except ValueError:
    print(f"Greška! {a} nije broj. Pokušajte ponovo.")
````

```{admonition} Napomena o opsegu delovanja try naredbe
:class: note 
Naredba `try` upravlja i izuzecima koje ne možete da specifikujete pa ih `except` ne može uhvatiti.
```

````python
a = [2]
try:
    if a <= 4:
        print(f"Broj {a} je manji ili jednak od 4")
except:
    print(f"Greška! Nešto nije u radu sa a={a}. Pokušajte ponovo.")
````

```{admonition} Rukovanje više različitih grešaka.
:class: note 
Naravno, `try` naredba može da ima više od jedne naredbe za izuzetak kako bi upravljala različitim izuzecima:
```

````python
def test_izuzeci(a):
    try:
        a = int(a)
        if a<=4:
            print(f"Broj {a} je manji ili jednak od 4")
    except TypeError:
        print(f"Greška! a={a} nije broj. Pokušajte ponovo.")
    except ValueError:
        print(f"Greška! Ne mogu da konvertujem a='{a}' u ceo broj. Pokušajte ponovo.")
    except:
        print("Greška! Neočekivana greška. Pokušajte ponovo.")
````

````python
test_izuzeci([2])
````

````python
test_izuzeci("s")
````

## Naredba `raise`

Drugi korisna stvar u Pajtonu je postojanje mogućnosti ukazivanja ne neki izuzetak u određenim okolnostima primenom naredbe `raise`. Na primer, ako treba da nam `p` ima vrednost koja je veća 0, tada možemo upotrebiti sledeći kod da pokrenemo izuzetak ako je `p` manje od 0. Program će prikazati naš izuzetak i zaustaviti izvršavanje:

````python
p  = -1
if p < 0:
    raise(Exception("Broj p predstavlja cenu i ne može biti negativan broj."))
````

## Provera tipova

U programskim jezicima kao što je Pajton ne postoji način da se osigura da korisnik vaše funkcije unosi promenljive tipa podataka koje očekujete. Funkcija `moje_sabiranje` dizajnirana je tako da sabira tri broja. Međutim, korisnik može uneti torke, liste, rečnike ili funkcije, od kojih će svaka uzrokovati različite nivoe problema. Možete zatražiti da vaš tip funkcije proveri ulazne promenljive pre nastavka i iznudi grešku pomoću funkcije greške.

:::{admonition} Primer 3.  
:class: tip

Napravite funkciju koja sabira tri broja `moje_sabiranje` tako da proverava da li su ulazne promenljive odgovarajućeg tipa. Ako bilo koja od ulaznih promenlivih nije odgovarajućeg tipa, funkcija bi trebala vratiti odgovarajuću grešku korisniku pomoću funkcije `raise`. Isprobajte svoju funkciju za pogrešne unosne argumente da biste proverili da li kod funkcioniše.
:::

````python
def moje_sabiranje(x, y, z):
    # provera tipova
    if isinstance(x, float) and isinstance(y, float) and isinstance(z, float):
        pass
    else:
        raise(TypeError('Ulazni argumenti moraju biti realni brojevi.'))
       
    return x + y + z
````

````python
moje_sabiranje(1.0, 12.1, 13.2)
````

````python
moje_sabiranje("1.0", 12.1, 13.2)
````

````python
moje_sabiranje(1, 12.1, 13.2)
````
