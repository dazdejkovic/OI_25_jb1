## Logičke promenljive

Naredna važna klasa osnovnih tipova podataka su *logičke promenljive*. One će uzimati dve vrednosti - `True` i `False` i sugerisaće nam istinitost određene tvrdnje.

````python
print(4>3)
print(type(4>3))
type(4>3) # ponovo primetite malo drugačiji ispis istog upita preko funkcije print() i poslednje unosa u komandnoj liniji
````

Logičke promenljive će nam puno koristiti. Dobra stvar je što Python njih konvertuje u cele brojeve bez dodatnih zahteva. Prebacivanje u cele brojeve, takođe daje očekivan rezultat 

````python
# Šta će biti ispis sledeće ćelije?

x = (4<3)
print(type(x))
x
# a - Error
# b - bool 
# c - int
````

````python
x = (4<3)
y = 23 * x
print(y, '\n', type(y))

# a - Error
# b - 0, int
````

````python
print(x)
print(int(False))
print(int(True))
````

### Logičke promenljive i brojevi

Zanimljiva stvar će se desiti ako u logičku promenljivu prebacimo neki broj. Nevezano sa tim da li je `int` ili `float`, logička promenljiva uzima vrednost `True` kad god broj nije nula.

````python
print(bool(10.2))
print(bool(0))

# a) Error
# b) True, False
````

### Relacioni operatori

Relacioni operatori porede dve vrednosti ili dva izraza i upoređuju njihovu vrednost u jednu logičku promenljivu. Koristimo sledeće operatore:
<div style="max-width: 600px; margin: auto;">
    
| Operator | Značenje      |
| :------: | :------------ |
| \==      | Jednako       |
| !=       | Nije jednako  |
| <        | Manje         |
| \>       | Veće          |
| <=       | Manje jednako |
| \>=      | Veće jednako  |

</div>

````python
(12-8) == 16 ** 0.5  
````

````python
x = 5
x != 12-8
````

~~~{admonition} Obratite pažnju na razliku između `=` i `==`
:class: warning
Jedna od čestih grešaka je mešanje ova dva operatora.

* Operator `=` služi da promenljivoj sa leve strane operatora **dodelimo** vrednost koja se nalazi sa desne strane
* Operator `==` služi da **proverimo** da li su vrednosti sa leve i desne strane jednake. 
~~~

### Logički operatori

Logički operatori porede dva logička izraza i kreiraju finalnu logičku promenljivu kao rezultat operacija. Logički operatori su:

* `and`
* `or`
* `not`

Slede razni primeri koda u kojima se pojavljuju logički operatori:

````python
print(True and False, True or False, not True)
````

````python
x = 7
(x != 3) or (x < 2)
````

````python
not x == 12
````

````python
not x = 12
````

## Promena tipa promenljivih

Lako se može desiti da imate neki broj koji je upisan kao string, a želite da izvršite neku računsku operaciju sa njim. Ako ne ukažete interpreteru da želite da radite sa brojem mogu se javiti greške.

Razmislite, šta će biti izlaz kod sledećeg koda za promenljivu `upit1`, a šta za `upit2`?

```python

s1 = '3'
s2 = '4'
upit1 = s1 + s2
upit2 = s1 * s2
upit3 = s1 * 4
```

Kako bismo nagovestili interpreteru da želimo da baratamo brojevima, iskoristićemo funkciju `int()` koja sugeriše da ta promenljiva treba da označava ceo broj

````python
s1 = '3'
s2 = '4'
s = int(s1) * int(s2)
print(s, type(s))
````

Slično, možemo prebaciti promenljive i u ostale formate pozivom sledećih funkcija:

- `int()` - u ceo broj
- `float()` - u broj sa pokretnim zarezom
- `str()` - u string
- `bool()` - u logičku promenljivu

Naravno, ako zahtevamo od interpretera da nam promeni tip promenljive, a on to nije u mogućnosti, javiće grešku

````python
s = 'abc'
print(int(s))
````

Suština programiranja je da od korisnika dobijemo niz ulaznih podataka sa kojima kasnije dalje radimo. Funkcija `input()` služi upravo za to. Korisnik unosi string koji kasnije učitavamo u neku promenljivu sa kojom nakon toga radimo normalno.

````python
x = input()
````

````python
print(x)
type(x)
````

Naravno, prazan unos može zbuniti korisnika koji nije svestan šta treba da unese. Zbog toga, unutar poziva funckije `input()` možemo uneti string kojem pojašnjavamo šta želimo od korisnika.

````python
x = input("Unesite broj između jedan i deset: ")

print("Vaš broj je " + x + ".")
````

~~~{admonition} Obratite pažnju
:class: warning
Funkcija `input()` promenljivoj dodeljuje string. Ako želite da kasnije ovaj broj koristite prilikom računa morate da upotrebite i funkciju `int()`:
~~~


````python
x = input("Unesite broj između jedan i deset: ")
#print(type(x))
print("Tip odmah prilikom unosa: ", type(x))
x = int(x)
#print(type(x))
print("Tip nakon modifikacije: " + str(type(x)))
````

Kako ne bismo imali nepotrebne linije koda, elegantno rešenje je da odmah napomenemo interpreteru da želimo ceo broj.

````python
x = int(input("Unesite broj: "))
print(x, type(x))
print("Kvadrat ovog broja je", x ** 2)
````
