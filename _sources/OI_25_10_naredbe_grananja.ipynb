# Naredbe grananja

Naredbe grananja koristimo da kreiramo strukturu odlučivanja, koja dozvoljava programu da ima više od jedne putanje izvršavanja.

## IF-ELSE naredbe

- Naredba grananja, **if-else**, ili **if** naredba kratko, je kod konstruisan da izvršava blok koda samo ako su ispunjeni određeni uslovi
- Ti uslovi su reprezentovani kao logički iskazi

```{admonition} **Pseudo kod jednostavne IF naredbe**
:class: important

~~~python
if logicki_izraz:
    blok koda
~~~

- Reč `if` je _ključna reč_
- Kada Pajton "vidi" neku `if` naredbu, on će proveriti da li je odgovarajući logički izraz tačan
    - Ako je tačan, izvršiće se blok koda koji sledi i završiti `if` naredbu
    - Ako nije tačan, blok koda koji sledi se preskače i završiti `if` naredbu   
```

Primetimo da je blok koda koji sledi `if` naredbu uvučen. Generalno, kad god se neka naredba u Pajtonu završava dvotačkom, sledi blok koda koji je uvučen i delovanje te naredbe je ograničeno na taj kod. Jedna od najčešćih grešaka prilikom kompajliranja u startu je da zaboravite da ispoštujete ovo pravilo. Neka vam ovo stalno bude na pameti. 

Blok koda mora biti uvučen za tačno četiri razmaka inače će prevodilac javiti grešku.

```{admonition} Primer 1.
:class: tip
Napisati program koji će uraditi sledeći zadatak: Ukoliko je vrednost promenljive veća od nule, promeniti joj vrednost u 1.
```

````python
x = float(input("Unesite jedan broj:"))
if x > 0:
    x = 1
print(x)
````

```{admonition} **Pseudo kod jednostavne if-else naredbe**
:class: important

~~~python
if logicki_izraz:
    blok koda 1
else:
    blok koda 2
~~~
```    

Kada Pajton "vidi" neku `if-else` naredbu, on će proveriti da li je odgovarajući logički izraz tačan.

 - Ako je tačan, izvršiće se prvi blok koda koji sledi i završiti `if-else` naredba
 - Ako nije tačan izvršiće se drugi blok koda i završiti `if-else` naredba.

~~~{admonition} Primer 2.
:class: tip

Korisnik unosi jedan broj. Napisati kod koji će izračunavati vrednosti funkcije: 
```{math}
f(x) = \begin{cases} 1 &\text{, ako je $x \ge 0$} \cr   -1 &\text{, ako je $x<0$} \end{cases}
```
~~~

````python
x = float(input("Unesite realni broj:"))
if x>=0:
    print(f"f({x}) = 1")
else:
    print(f"f({x}) = -1")
# Razlikujte značenje za dva f koja se pojavljuju u ovom primeru.
````

```{admonition} Primer 3.
:class: tip
Napisati program koji će uraditi sledeći zadatak: Korisnik unosi string. Ukoliko string počinje malim slovom, izmeniti ga tako da počinje velikim slovom. Ukoliko string počinje velikim slovom, izmeniti ga da počinje malim.
```

````python
s = input("Unesite string:")
if s[0] == s.lower()[0]:
    s = s.capitalize() # Ne može samo s.capitalize() jer metoda onda neće promeniti sam string
else:
    s = s.lower()
print(s)
````

```{admonition} Pseudo kod proširene if-else naredbe:
:class: important
~~~python
if logicki izraz P:
    blok koda 1
elif logicki izraz Q:
    blok koda 2
elif logicki izraz R:
    blok koda 3
else:
    blok koda 4
~~~
```
    
U prethodnom pseudokodu Pajton će: 
-  Prvo proveriti da li je `P` tačno. Ako je `P` tačno, izvršiće se blok koda 1 i završiće se `if-else` naredba. Drugim rečima, Pajton *neće* proveravati ostale iskaze kad pronađe prvi tačan iskaz.
- Ako je `P` netačno, tada će Pajton proveriti da li je `Q` tačno. Ako je `Q` tačno, izvršava se blok koda 2 i završava `if-else` naredba. 
- Ako su `P` i `Q` netačni, tada će Pajton proveriti da li je `R` tačno. Ako je `R` tačno, izvršava se blok koda 3 i završava `if-else` naredba. 
- I konačno, ako su `P`, `Q` i `R` netačni, izvršava se blok koda 4 i završava `if-else` naredba.

```{admonition} Napomena:    
:class: caution
- Možemo da imamo proizvoljan broj `elif` naredbi (ili nijednu) dok god imamo barem jednu naredbu `if` (prva naredba)
- Nije neophodna `else` naredba, ali možemo da imamo najviše jednu naredbu`else`    
```    

~~~{admonition} Primer 4.
:class: tip
Korisnik unosi dve vrednosti `temp` i `trazena_temp`. Napisati kod koji će poslati poruku  `"Toplo"` ako je `temp` za barem 5 stepeni veća od `trazena_temp`, `"Hladno"` ako je barem `temp` za 5 stepeni manja od `trazena_temp`, i `"Normal"` u ostalim slučajevima. 
~~~

````python
temp = float(input("Unesite trenutnu temperaturu:"))
trazena_temp = float(input("Unesite traženu temperaturu:"))
if temp - trazena_temp > 5:
    status = "Toplo"
elif temp - trazena_temp < -5:
    status = "Hladno"
else:
    status = "Normal"
print(status)
````

## Ugneždene naredbe

- Naredba je **ugneždena** ako se cela nalazi unutar druge naredbe istog tipa kao i ona sama.
- Na primer, **ugneždena if naredba** je neka if naredba koja se cela nalazi unutar klauzule druge if naredbe

~~~{admonition} Kviz 3.
:class: hint

Razmislite o tome šta će se desiti pošto se izvrši sledeći kod. Koji su sve mogući ishodi u zavisnosti od vrednosti
x i y? Šta će biti odgovori za sledeća dva para brojeva: (5,-2) i (1,2)?
```python
x = 5
y = -2

if x > 2:
    if y < 2:
        z = x + y
    else:
        z = x - y
else:
    if y > 2:
        z = x * y
    else:

~~~{admonition} Primer 5. 
:class: tip 
Napišite program koja uzima broj, `r`, i string `mera`, kao ulazne argumente. Pretpostavite da 
je `r` pozitivan, i da je `mera` ili string `'povrsina'` ili string `'obim'`. Program treba da izračunava površinu kruga

## Ternarni operatori

**Ternarni operatori** ili **uslovne naredbe** nam obezbeđuju upotrebu koda u jednoj liniji da izračunamo prvi izraz ako je 
uslov istinit; u suprotnom izračunava se drugi izraz.

```{admonition} Pseudo kod za ternarnog operatora 
:class: important

~~~python 
izraz_ako_istina if uslov else izraz_ako_laz 
~~~
    


````python
status_student = False
osoba = "student" if status_student else 'nije student'
print(osoba)
````

Ternarni operatori omogućavaju jednostavan način za grananje i mogu učiniti naš kod jasnijim. Videćemo uskoro 
njihovu ulogu u objašnjavajućim listama (*comprehension list*).
