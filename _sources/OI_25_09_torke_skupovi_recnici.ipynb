# Torke, skupovi i rečnici



Objekti kao što su torke, rečnici i skupovi su iterabilni objekti. I mada to sada nije toliko bitno radi se o hashabilnim objektima, odnosno o objektima koji se mogu pretvoriti u stabilan celobrojni kod (hash) koji se ne menja tokom njihovog “života”.

## Tuples (n-torke, torke)

Torka je uređena i nepromenljiva (immutable) kolekcija elemenata. Pišu se u zagradama. Elementi imaju redosled i pristupamo im preko indeksa. Primer:
```
t = (1, 3.14, "Python")
```
Torke se koriste za podatke koji treba da ostanu nepromenljivi, ili kao povratne vrednosti funkcija.

### Definisanje torke i osnovne operacije

````python
torka = 2, 5.6, 34, "element"

tup = (2, 5, 63)
````

````python
print(type(tup))
print(type(torka))
#print(torka)
````

Isecanje radi isto kao kod stringova i lista:

````python
torka[2:4]
````

Kao što smo rekli torke su nemutabilni objekti pa će bilo koji pokušaj izmene njenog sadržaja dovesti do greškeČ

````python
torka[4] = 45
````

Zamena vrednosti dva broja se najlakše vrši koristeći torke kao strukturu podataka u pozadini. Sledeći kod ilustruje pristup većine programskih jezika pomoću dodatne promenljive:

````python
a = 6
b = 7
pom = a
a = b
b = pom
print(a,b)
````

U Pajtonu sa torkama to je vrlo jednostavno:

````python
x, y = 5, 10
print(x,y)
x, y = y, x
print(x, y)
````

Iako su torke nemutabilni objekti, interesantno je da ako torka bude sastavljena od mutabilnih objekata, njih i dalje možete menjati:

````python
tup = ([2, 4, 5], ['a', 'b', 't'])
tup[1].append('dodatak')
tup
````

Ugrađenim komandama `list` i `tuple` možemo listu prebaciti u torku, i obratno:

````python
tup = (4, 6, 7)
l = list(tup)
print(l, type(l))
t1 = tuple(l)
t1
````

Izbor tipa podataka, naravno zavisi od problema. Torke su zgodnije ako ne želimo da dozvolimo izmenu podataka. Torke uzimaju manje memorijskog prostora, što takođe može biti zgodno.

````python
lista = [1, 3, 6.8, .5, "jos nesto"]
torka = (1, 3, 6.8, .5, "jos nesto")

import sys 
print(sys.getsizeof(lista))
print(sys.getsizeof(torka))
````

## Skupovi

Skup je kolekcija jedinstvenih elemenata bez definisanog redosleda. Duplikati se automatski uklanjaju. Elementi nisu indeksirani i omogućavaju brze operacije teorije skupova. Pišu se unutar `{ }`.

````python
skup = {'jabuka', 'banana', 'kivi'}
print(skup, type(skup))
````

Tip greške u sledećem kodu vrlo lepo pokazuje o činjenicu da elementi nisu indeksirani:

````python
skup[1] 
````

Skupovi su mutabilni. Metoda `.add` koristi se za dodavanje novog elementa u skup:

````python
s={'a', 'b', 'c', 'd', 'e'}
s.add("x")
s
````

Budite oprezni oko dodavanja elemenata u skup. Metoda `.add()` dodaje element u integralnom obliku, dok biste sa `set("neki string")` dobili svako slovo tog elementa u skupu:

````python
skup = set("neki string")
skup
````

````python
skup.add('drugi string')
skup
````

Za skupove postoje sve logičke operacije koje postoje i za regularne operacije sa skupovima:

````python
s & {"a","b",3} # Presek 
````

````python
s | {"g", "h"} # Unija
````

````python
s - {"a","b"} # Razlika skupova
````

## Rečnici

Rečnici su jedan od važnijih tipova podataka. Predstavljaju neuređenu, promenljivu kolekciju, koja skladišti parove objekata `kluč:vrednost` (tj. `key:value`). 
- Kjuč je nemutabilni objekat (string najčešće)
- Za razliku od listi i ostalih sekvencijalnih tipova podataka, gde se elementu pristupa preko indeksa pozicije, rečnici omogućavaju pristup preko ključa
- Rečnik formiramo unutar vitičastih zagrada, nove pojmove odvajamo zarezom

````python
d = {'Petar': 9, # Jedno ispod drugog samo radi preglednosti
     'Milica': 6,
     'Marko': 9 # kljucevi moraju biti razliciti, ali vrednosti ne
    }
print(type(d))
#Ovde su ključevi imena. Da pravimo nesto ozbiljnije oko ocena studenata šta bi bio smisleniji ključ?
````

````python
d['Petar']
````

````python
d['Milica'] = 7 #vrednosti mozemo da menjamo
````

:::{admonition} Važno
:class: important

Ako kreiramo promenljivu `a = {}` ona može biti ili prazan rečnik ili prazan skup. Ipak, zbog značaja, na ovaj način formiramo prazan rečnik uvek. Ako želimo prazan skup da napravimo moramo to uraditi na sledeći način `X = set()` 
:::

````python
a = {}
type(a)
````

### Osnovne operacije sa rečnicima
    
Unos novih elemenata u rečnik vrlo je jednostavan i ni ne zahteva pozivanje metoda:

````python
glavni_gradovi = {}
glavni_gradovi['Srbija'] = 'Beograd'
glavni_gradovi['Rusija'] = 'Moskva'
````

Za brisanje iz rečnika, kao i kod listi, koristimo komandu `del`:

````python
del glavni_gradovi['Rusija']
glavni_gradovi
````

Ispravno struktuirane podatke moguće je pretvoriti u rečnik ugrađenom funkcijom `dict`

````python
recept = dict([('jaja (kom)', 2), ('brasno (gr)', 100), ('voda (l)', 0.5)])
print(recept, "\n", type(recept))
````

### Osnovne metode na objektima tipa rečnik

Metode `items()`, `keys()` i `values()` omogućavaju da pogledamo ključeve i vrednosti nekog rečnika:

````python
glavni_gradovi.items() 
````

````python
print(glavni_gradovi.keys())
# Ne zaboravite da stalno imate na umu šta je tip podataka koji dobijamo     
print(type(glavni_gradovi.keys()))
````

````python
print(glavni_gradovi.values())
````

Ono što u ovom trenutku treba da zapamtite da sve ove funkcije generišu iterabine objekte, ali to nisu liste ni taplovi.
