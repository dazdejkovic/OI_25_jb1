# Osnovne napomene

## Ciljevi

- Učenje programiranja u Python-u studenata ekonomije koji pre toga nisu imali nikakava iskustva u programiranju
- Uvod u različite analitičke alate koji su korisni u rešavanju ekonomskih i biznis problema
- Poslovne simulacije
- Uvod u osnovne optimizacione tehnike korišćenjem Python-a

## Sadržaj kursa

- Osnove Python-a
- Numpy - paket za numerička izračunavanja
- Pandas - paket za rad sa podacima
- Matplotlib - paket za grafički prikaz podataka
- MIP - paket za rešavanje optimizacionih problema
- Sympy - paket za simbolička izračunavanja (ako stignemo)

Kroz primere, zadatke i probleme iz ekonomije

## Zašto Python ?

- Trenutno ubedljivo najpopularniji "data science" programski jezik uz R
- Razlika u odnosu na R: Python je jezik opšte namene koji nas vodi ka velikom broju primena
- Besplatan je (open source)
- Postoji velika zajednica korisnika koja ga podržava, ostavljajući veliki broj uputstava, biblioteka i podataka javno dostupnim
- Jednostavan je za učenje (user friendly)
